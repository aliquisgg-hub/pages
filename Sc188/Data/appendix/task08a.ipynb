```python
# pandas ライブラリをインポートする (使える状態にする)
import pandas as pd

# 完了確認のメッセージ
print("完了!")
```

```python
# ファイル "task08.xlsx" のワークシート "販売" を変数 hanbai に読み込む
# （「dataframe の操作」の「Excel ファイルの読み込み」を参考に）
hanbai = pd.read_excel("task08.xlsx", sheet_name="販売")
hanbai
```

```python
# ファイル "task08.xlsx" のワークシート "レシート" を変数 receipt に読み込む
# （「dataframe の操作」の「Excel ファイルの読み込み」を参考に）
receipt = pd.read_excel("task08.xlsx", sheet_name="レシート")
receipt
```

```python
# ファイル "task08.xlsx" のワークシート "商品" を変数 shohin に読み込む
# （「dataframe の操作」の「Excel ファイルの読み込み」を参考に）
shohin = pd.read_excel("task08.xlsx", sheet_name="商品")
shohin
```

```python
# shohin を見ると，m001 と m004 は重複しているので，hanbai の "商品コード" で "m004" を "m001" に置き換える
# （「データクレンジング」の「表記のゆらぎの統一」を参考に）
hanbai["商品コード"] = hanbai["商品コード"].str.replace("m004", "m001")
hanbai
```

```python
# (課題には影響しないが) shohin から "商品コード" が "m004" の行を削除するには以下のようにする
shohin = shohin[shohin["商品コード"]!="m004"]
shohin
```

```python
# receipt の "性別" の欠損値を最頻値で補う
# （「データクレンジング」の「欠損値の補充」を参考に）
mode_of_gender = receipt["性別"].mode()[0]    # 性別の最頻値
receipt["性別"] = receipt["性別"].fillna(mode_of_gender)
receipt
```

```python
# receipt の "年齢" の欠損値を平均値(の四捨五入)で補う
# （「データクレンジング」の「欠損値の補充」を参考に）
mean_of_age = round(receipt["年齢"].mean())   # 顧客の年齢の平均値
receipt["年齢"] = receipt["年齢"].fillna(mean_of_age)
receipt
```

```python
# "レシート番号" を介して hanbai に receipt を結合して joined1 に代入する
# （「dataframe の操作」の「結合 (join)」を参考に）
joined1 = pd.merge(hanbai, receipt, on="レシート番号")
joined1
```

```python
# "商品コード" を介して joined1 に shohin を結合し，joined に代入する
# （「dataframe の操作」の「結合 (join)」を参考に）
joined = pd.merge(joined1, shohin, on="商品コード")
joined
```

```python
# joined の各行を，"日時" の順に並べ替えて sorted に代入する
# （「dataframe の操作」の「行の並べ替え」を参考に）
sorted = joined.sort_values("日時")
sorted
```

```python
# sorted のインデックスを振り直して final に代入する
# （「dataframe の操作」の「インデックス (通し番号) の振り直し」を参考に）
final = sorted.reset_index(drop=True)
final
```

```python
# 出来上がった表 (final) を，CSVファイル "task08final.csv" に保存する
# （「dataframe の操作」の「CSVファイルへの保存」を参考に）
final.to_csv("task08final.csv", index=False, encoding="utf_8_sig")

# 確認のため，改めて data08final.csv 読み込んで (dataframe に変換して) 表示する
pd.read_csv("task08final.csv")
```
