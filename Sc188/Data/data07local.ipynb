```python
# カリフォルニア州の各地域内の住宅価格のデータセット
# https://inria.github.io/scikit-learn-mooc/python_scripts/datasets_california_housing.html

# データの読み込み
import pandas as pd
cal_housing = pd.read_excel("data06.xlsx", sheet_name="住宅価格", usecols="A:I")

# 目的変数（MedHouseVal）を Series型 に変換して変数 x に入れる
x = cal_housing["MedHouseVal"]

# 目的変数 (MedHouseVal) は各地域の住宅価格の中央値で，十万ドル ($100,000) 単位で表されている

# データ (x) の表示（最初と最後が表示される）
display(x)
```

```python
# MedHouseVal の基本統計量とヒストグラムの表示
import matplotlib.pyplot as plt
b = 50                  # ヒストグラムの階級数
plt.hist(x, bins=b)
print("平均：", x.mean())
print("分散：", x.var(ddof=0))
```

```python
# ランダム・サンプリングを繰り返す（30秒以上かかるかも）
n = 1000    # サンプリング時のサンプル数
r = 50000   # サンプリングの反復回数

y = []
for i in range(0, r):
    y.append(x.sample(n).mean())    # 標本平均のデータを取る
y = pd.Series(y)    # r個の標本平均を Series型 のデータに変換して変数 y に入れる

# データ y の表示（最初と最後が表示される）
display(y)
```

```python
# r個の標本平均の基本統計量とヒストグラムの表示
b = 50                  # ヒストグラムの階級数
plt.hist(y, bins=b)
print("平均：", y.mean())
print("分散：", y.var(ddof=0))
```

```python
# 母集団と標本平均の，平均と分散の比較
print("母集団の平均-標本平均の平均：", x.mean()-y.mean())
print("母集団の分散/標本平均の分散：", x.var(ddof=0)/y.var(ddof=0))
```
