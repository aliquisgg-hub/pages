## 日本語ワードクラウドの作成

```python
import pandas as pd

# Janome (https://mocobeta.github.io/janome/) を
# 用いて形態素解析した結果を読み込む
tokens = pd.read_csv("sangetsuki.csv")
#tokens = pd.read_csv("mojika.csv")
#tokens = pd.read_csv("meijinden.csv")
tokens
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt

# 名詞，動詞，形容詞 を抽出
ind = tokens[tokens["品詞"]=="名詞"]
ind = pd.concat([ind, tokens[tokens["品詞"]=="動詞"]])
ind = pd.concat([ind, tokens[tokens["品詞"]=="形容詞"]])
ind
```

```python
words = " ".join(list(ind["原形"]))

# 一部の語を除去
rm_words = [
    "ある", "いる", "する", "せる", "なる", "これ", "それ", "あれ", "どれ", 
    "この", "その", "あの", "どの", "もの", "こと", "よう", "れる", "の",
    "ない"
]
for w in rm_words:
    words = words.replace(f" {w} ", " ")

words
```

```python
# ワードクラウドの生成表示
## 日本語のフォントファイルを指定する
wc = WordCloud(font_path="/c:/Windows/Fonts/YuGothM.ttc", width=500, height=500)
wc.generate(words)
plt.figure(dpi=150)     # dpi の数字を大きくすると，大きく表示
plt.imshow(wc)
plt.axis("off")         # 軸目盛りの非表示
plt.show()
```
