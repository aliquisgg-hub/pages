```python
# あらかじめ，左のサイドバーの「ファイル」で，読み込むファイルをアップロードしておく
f = open("sangetsuki.txt")
text = f.read()
f.close()
text = text.translate(str.maketrans("\n\u3000", "  ", "\ufeff")) # これはなくてもよいかも
text
```

```python
tokenizer = Tokenizer()
tlist = tokenizer.tokenize(text)
tab = []

for t in tlist:
    if t.extra==None:
        t.extra = (t.part_of_speech, '*', '*', t.surface, t.surface, t.surface)
    tab.append([t.surface] + t.part_of_speech.split(',') + list(t.extra[1:]))

df = pd.DataFrame(tab)
df.columns = ['表層形', '品詞', '品詞細分類1', '品詞細分類2', '品詞細分類3', '活用型', '活用形', '原形', '読み', '発音']
df
```

```python
df.to_csv('sangetsuki.csv', index=False, encoding='utf_8_sig')
```
