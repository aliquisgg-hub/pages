## 日数計算の関数

```python
def calcDate(year, month, day):
    mTable = [0, 31, 59, 90, 120, 151, 181, 212, 243, 273, 304, 334]
    mTableU = [0, 31, 60, 91, 121, 152, 182, 213, 244, 274, 305, 335]
    if year % 4 == 0:
        d = mTableU[month-1] + day
    else:
        d = mTable[month-1] + day
    y = year - 1901
    return y * 365 + y // 4 + d

print(calcDate(1901, 1, 1))     # ⇒ 1
print(calcDate(2020, 2, 28))    # ⇒ 43523
print(calcDate(2020, 3, 1))     # ⇒ 43525
print(calcDate(2020, 12, 31))   # ⇒ 43830
print(calcDate(2021, 1, 1))     # ⇒ 43831
```

## ソーティング（並べ替え）の関数A

```python
def swap(l, i, j):
    l[j], l[i] = l[i], l[j]

def indexOfMax(l):
    maxi = 0                # 最初は (暫定的に) l[0] が最大と仮定する
    i = 1                   # i を 1 から
    while i < len(l):       # … len(l)-1 まで変えながら反復
        if l[i] > l[maxi]:  # もし l[i] が l[maxi]（それまでの暫定的な最大値）より大きいとき
            maxi = i        # l[i] を (暫定的な) 最大値として maxi を iの値 で置き換える
        i = i + 1
    return maxi             # 戻り値は maxi

def mySortA(l):
    n = len(l)
    while n > 0:
        m = indexOfMax(l[:n])
        if l[m] > l[n-1]:
            swap(l, m, n-1)
        n = n - 1

l1 = [1, -4, 2, -8, 5, -7]
mySortA(l1)
print(l1)                        # ⇒ [-8, -7, -4, 1, 2, 5]
l2 = [1, 4, 6, 2, 8, 5, 3, 7]
mySortA(l2)
print(l2)                        # ⇒ [1, 2, 3, 4, 5, 6, 7, 8]
```

## ソーティング（並べ替え）の関数B

```python
def mySortB(l):
    m = []
    for x in l:
        i = 0
        while i < len(m):
            if x < m[i]:
                break
            i = i + 1
        m.insert(i, x)
    return m

print(mySortB([1, -4, 2, -8, 5, -7]))     # ⇒ [-8, -7, -4, 1, 2, 5]
print(mySortB([1, 4, 6, 2, 8, 5, 3, 7]))  # ⇒ [1, 2, 3, 4, 5, 6, 7, 8]
```
