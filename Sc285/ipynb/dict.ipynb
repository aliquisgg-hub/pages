## 辞書の操作

```python
dict_obj = {}
dict_obj["dog"] = "犬"

print(dict_obj["dog"])
```

```python
dict_obj = {"dog": "犬"}
print("修正前:", dict_obj["dog"])

dict_obj["dog"] = "わんこ"
print("修正後:", dict_obj["dog"])
```

```python
dict_obj = {"dog": "犬", "cat":"猫"}

del dict_obj["dog"]
print(dict_obj)
```

```python
dict_obj = {"dog": "犬", "cat":"猫"}

del dict_obj["penguin"]
```

```python
dict_obj = {"dog": "犬", "cat":"猫"}
print(dict_obj)
print(len(dict_obj))  # dict_objの要素数を求める
```

```python
dict_obj["penguin"] = "ぺんぎん"  # dict_objに要素を追加する
print(dict_obj)
print(len(dict_obj))  # dict_objの要素数を求める
```

```python
del dict_obj["cat"]  # dict_objの要素を削除
print(dict_obj)
print(len(dict_obj))  # dict_objの要素数を求める
```

## in 演算子

```python
english_words = {"apple": "りんご", "orange": "みかん", "peach": "もも"}
print("apple" in english_words)
```

```python
english_words = {"apple": "りんご", "orange": "みかん", "peach": "もも"}
print("dog" in english_words)
```

```python
english_words = {"apple": "りんご", "orange": "みかん", "peach": "もも"}

key = input("英単語を入力してください")
print("入力:", key)
if key in english_words:
    # キーが english_words に登録されている
    print(english_words[key])
else:
    # キーが english_words に登録されていない
    print("登録されていません")
```
