## 計算練習プログラムのための部品となる関数を定義する

- **makeProb()**: 計算練習問題を1つ作る関数
- **askProb(prob)**: 問題を出題して解答を入力させ，問題に結果（解答と判定結果）を追加して返す関数
- **showResult(result)**: askProb 関数の戻り値（問題と結果）を使って，応答メッセージを表示する関数
- **makeProbList(n)**: n個の問題を生成して，そのリストを返す関数
- **askProbList(probList)**: 与えられた問題リストの問題を順に出題し，出題ごとに解答を入力させ，「問題と結果」のリストを返す関数
- **showResultList(resultList)**: 与えられた「問題と結果」のリストをまとめて表示する関数

### **makeProb()**: <br>　　計算練習問題を1つ作る関数
- 引数はなし
- 戻り値は，作成した問題とする．
  - `(演算子, 左の数, 右の数, 正解)` という形式のタプルで返すものとする．
  - `演算子` は `"+"`, `"-"` などの文字列で表すものとする．
  - `左の数`, `右の数`, `正解` は数値で表すものとする．
  - 例えば，5 + 3 なら `("+", 5, 3, 8)` というタプルを戻り値として返すことになる．
- [random.randint 関数](https://docs.python.org/ja/3.8/library/random.html?highlight=randint#random.randint)を使ってランダムに問題を作成する．
- さしあたって，1～10 の自然数の足し算の問題を作成するものとする．

〔最終課題へのヒント〕
- 計算練習プログラムを改変して，問題の種類を変えるときは，この関数を修正すればよい．
- 異なった種類の問題（足し算と引き算など）を混在させる場合は，その種類ごとに異なった名前で）この makeProb のような関数を作成して使い分けてもよいし，
一つの makeProb() 関数の中で乱数で場合分けをして，異なった種類の問題を作り分けてもよい．
　 

````python
import random

def makeProb():
    op = "+"
    l = random.randint(1, 10)
    r = random.randint(1, 10)
    sol = l + r
    return (op, l, r, sol)

makeProb()   # 例えば ("+", 5, 3, 8) のような形のタプルを返すはず．
````

### **askProb(prob)**: <br>　　問題を出題して解答を入力させ，問題に結果（解答と判定結果）を追加して返す関数
- 引数 prob には，出題する問題を makeProb 関数の戻り値と同じ形式のタプルで与える．
  - `(演算子, 左の数, 右の数, 正解)` という形式．
  - 引数のタプルの内容の各々を個別に取り出すには，アンパック（複数の変数への代入）を用いるとよい．
- 戻り値は，解答とその解答の正誤を判定した結果を引数のタプルに追加したタプル `(演算子, 左の数, 右の数, 正解, 解答, 判定結果)` とする．
  - `解答` は解答者が回答した値（数値）とする．
  - `判定結果` は，解答が正解(と等しい)ならば `True` そうでなければ `False` とする．
  - 例えば，`("+", 2, 5, 7)` というタプルが引数に与えられ，7 と解答すれば，`("+", 2, 5, 7, 7, True)` というタプルを返すことになる．

  ※ タプルなので，`.append` や `.extend` は使えない
- 出題と解答の入力には，input 関数を用いるとよい．

※ input 関数で入力された値（戻り値）は文字列なので，数値に変換するのを忘れないように．

````python
def askProb(prob):
    op, l, r, sol = prob    # アンパック
    ans = int(input(f"{l} + {r} = ? "))
    judge = (ans==sol)
    return (op, l, r, sol, ans, judge)

prob = ("+", 5, 3, 8)
askProb(prob)  # 例えば 8 と解答すると ("+", 5, 3, 8, 8, True) というタプルを，7 と解答すると ("+", 5, 3, 8, 7, False) というタプルを返すはず．
````

〔ヒント〕
- 上の askProb 関数や，下の showResult 関数などでは，[フォーマット済み文字列リテラル](https://docs.python.org/ja/3.8/reference/lexical_analysis.html#f-strings)を使うと便利かも

````python
# フォーマット済み文字列リテラル …… f を前につけた文字列リテラル（引用符は " でも ' でもよい）
#   文字列中に {} で括って式を書くと，その部分が式の値で置き換えられる

print(f"1 + 2 = {1+2}")
x, y = 1, 2
print(f'{x} + {y} = {x+y}')
````

### **showResult(result)**: <br>　　askProb 関数の戻り値（問題と結果）を使って，応答メッセージを表示する関数
- 引数 result には，askProb 関数の戻り値と同じ `(演算子, 左の数, 右の数, 正解, 解答, 判定結果)` という形式のタプル（問題と結果）を与えるものとする．
- 戻り値はなし．
- 応答メッセージの表示は，例えば以下のような形式が考えられる（適当にアレンジしてもよい）．
  - 正解の場合:
    ```plaintext
    問題：7+5，解答：12 － ○
    ```
  - 不正解の場合:
    ```plaintext
    問題：8+5，解答：15 － ×，正解は13です
    ```

````python
def showResult(result):
    op, l, r, sol, ans, judge = result  # アンパック
    if judge:
        print(f"問題：{l}+{r}，解答：{ans} － ○")
    else:
        print(f"問題：{l}+{r}，解答：{ans} － ×，正解は{sol}です")

showResult(('+', 7, 5, 12, 12, True))
showResult(('+', 8, 5, 13, 15, False))
````

### **makeProbList(n)**: <br>　　n個の問題を生成して，そのリストを返す関数
- 引数 n には，問題数を数値で与える．
- 戻り値は，「makeProb 関数の戻り値と同じ形式の問題表現」のリスト（長さ n）を返すものとする．
  - `[(演算子, 左の数, 右の数, 正解), ...]` という形となる．
  - 例えば，n が 3 なら，`[('+', 7, 5, 12), ('+', 8, 5, 13), ('+', 3, 8, 11)]` のような形のリストを返すはずである．
- 個々の問題の生成には，makeProb 関数を用いるとよい．
- 問題の重複は気にしないで n 問作るなら， `for _ in range(n):` が便利？ 
  - リストの内包表記を使えば，もっと簡単かも
- リスト（の末尾）への追加は，[.append](https://docs.python.org/ja/3.8/tutorial/introduction.html?highlight=list%20append#lists) メソッドを使うのが簡単．

````python
def makeProbList(n):
    pl =[]
    for _ in range(n):
        pl.append(makeProb())
    return pl

makeProbList(3)     # 例えば [('+', 7, 5, 12), ('+', 8, 5, 13), ('+', 3, 8, 11)] のような形のリストを返すはず．
````

〔最終課題へのヒント〕
- 同じ問題が重複しない問題リストを作成するには，この関数で工夫するとよい．
  - 「問題作成をn回反復する」ではなく，「問題のリストの長さがnになるまで反復する」という考えたほうがやりやすいかも．
    - すでにリストにある問題はリストに追加しないとか ……
- ある問題と同じ問題が，ある問題のリストに含まれているかどうかの判定は，[in 演算子 や not in 演算子](https://docs.python.org/ja/3.8/reference/expressions.html#in)を使うと便利かも

````python
def makeProb3():    # テスト用
    op = "+"
    l = random.randint(1, 3)
    r = random.randint(1, 3)
    sol = l + r
    return (op, l, r, sol)

def makeProbList(n):
    pl =[]
    while len(pl)<n:
        prob = makeProb3()  # テスト用にmakeProb3関数を呼び出している
        if prob not in pl:
            pl.append(prob)     # prob が pl に含まれていな場合のみ，この行を実行する？
    return pl

makeProbList(9)
````

### **askProbList(probList)**: <br>　　与えられた問題リストの問題を順に出題し，出題ごとに解答を入力させ，「問題と結果」のリストを返す関数
- 引数 probList には，makeProbList 関数の戻り値と同じ形式の，問題表現のリストを与えるものとする．
  - `[(演算子, 左の数, 右の数, 正解), ...]` という形．
- 個々の問題の出題と解答の入力には，for in と askProb 関数を用いるとよい．
- 戻り値は，askProb 関数の戻り値と同じ形式のタプル（問題と結果）のリストを返すものとする．
  - 要するに，`(演算子, 左の数, 右の数, 正解, 解答, 判定結果)` という形のタプルのリストを返すということ．

````python
def askProbList(probList):
    rl = []
    for prob in probList:   # probList の問題の一つ一つについて，以下を反復
        rl.append(askProb(prob))
    return rl

probList = [('+', 7, 5, 12), ('+', 8, 5, 13), ('+', 3, 8, 11)]
askProbList(probList)   # 例えば，[('+', 7, 5, 12, 12, True), ('+', 8, 5, 13，15, False), ('+', 3, 8, 11, 11, True)] のような形のリストを返すはず．
````

### **showResultList(resultList)**: <br>　　与えられた「問題と結果」のリストをまとめて表示する関数
- 引数 resultList には，askProbList 関数の戻り値と同じ形式の「問題と結果のタプルのリスト」を与えるものとする．
  - `[(演算子, 左の数, 右の数, 正解, 解答, 判定結果), ...]` という形．
- 戻り値はなし．
- 表示は，例えば以下のような形式が考えられる（適当にアレンジしてもよい）．
  ```plaintext
  問題：7 + 5，解答：12 ― ○
  問題：8 + 5，解答：15 ― ×，正解は13です問題：3 + 8，解答：11 ― ○
  ```

※ showResult 関数を用いてもよい．

〔最終課題へのヒント〕
- 正解数を数えるには，判定結果が True のタプルの数を数えればよい

````python
def showResultList(resultList):
    for result in resultList:   # resultList の「問題と結果」の一つ一つについて，以下を反復
        showResult(result)

testList = [
    ('+', 7, 5, 12, 12, True),
    ('+', 8, 5, 13, 15, False),
    ('+', 3, 8, 11, 11, True)]

showResultList(testList)
````
