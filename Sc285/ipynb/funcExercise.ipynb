### 例題2

- 身長(m)と体重(kg)の引数，各々 height，weight を取り，値から BMI を計算し，その値を戻り値として返す関数 bmi(height, weight) を定義してみよう．
- 戻り値には return 文を用いる．
- BMI の計算式は以下の通り

  $$
  BMI = \frac{体重}{身長 \times 身長}
  $$

```python


print(bmi(1.6, 54))             # ⇒ 21.09375
print(bmi(1.6, 54) * 1.6 * 1.6) # ⇒ 54.0
```

### 例題4

- 非負整数の引数 n と k (k≦n) を取り，二項係数 ${}_n{\mathrm C}_{\,k}$ を計算して，その値を戻り値として返す関数 comb(n, k) を定義してみよう．
- 例題3で定義した fact(n) を用いて作ること．
- math.comb は使わないで！
- 戻り値には return 文を用いる．
- 二項係数の計算式は以下の通り（$n!$ は $n$ の階乗を表す）
  $$
  \displaystyle{}_n{\mathrm C}_{\,k} = \frac{n!}{k!\,(n-k)!}
  $$

```python


print(comb(5,2))    # ⇒ 10.0
```
