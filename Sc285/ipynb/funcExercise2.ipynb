# 「関数の定義」の例題

### 例題1

- 数値の引数 x を取り，その引数の平方根を表示する関数 printSqrt(x) を定義してみよう．
- 負の数が入力された場合は，その数の負号を取って正の数にしたものの平方根を用いて，その前後に 0 + と i を連結する．  
  **例：**  
  　-4　⇒　0 + 2.0 i  
  　-2　⇒　0 + 1.4142135623730951 i
- 平方根は math.sqrt 関数を用いて計算をする．
- 結果の出力には print 関数を用いる．

ヒント：
- 先週の sqrt.py を参考に ...

```python
def printSqrt(x):
    import math
    if x>=0:
        print(math.sqrt(x))
    else:
        print("0 +", math.sqrt(-x), "i")

printSqrt(2)
printSqrt(-3)
printSqrt(0)
```

### 例題2

- 身長(m)と体重(kg)の引数，各々 height，weight を取り，値から BMI を計算し，その値を戻り値として返す関数 bmi(height, weight) を定義してみよう．
- 戻り値には return 文を用いる．
- BMI の計算式は以下の通り

  $$
  BMI = \frac{体重}{身長 \times 身長}
  $$

```python
def bmi(height, weight):
    b = weight/height/height
    return b

print(bmi(1.6, 54))             # ⇒ 21.09375
print(bmi(1.6, 54) * 1.6 * 1.6) # ⇒ 54.0
```

```python
# 別解

def bmi(height, weight):
    return weight/height/height

print(bmi(1.6, 54))             # ⇒ 21.09375
print(bmi(1.6, 54) * 1.6 * 1.6) # ⇒ 54.0
```

### 例題3

- 非負整数の引数 n を取り，その引数の階乗を計算し，その値を戻り値として返す関数 fact(n) を定義してみよう．
- math.factorial は使わないで！
- 戻り値には return 文を用いる．

ヒント：
- 先週の fact.py を参考に ...

```python
def fact(n):
    f = 1

    i = 1
    while i<=n:
        f = f * i
        i = i + 1

    return f

print(fact(5))                  # ⇒ 120
print(fact(5)/fact(2)/fact(3))  # ⇒ 10.0

print(fact(0))  # 0の階乗も問題なく動作 (場合分けの必要はない)
```

```python
# 別解1

def fact(n):
    f = 1

    while n>0:
        f = f * n
        n = n - 1
    return f

print(fact(5))                  # ⇒ 120
print(fact(5)/fact(2)/fact(3))  # ⇒ 10.0

print(fact(0))  # 0の階乗も問題なく動作 (場合分けの必要はない)
```

```python
# 別解2

def fact(n):
    if n==0:
        return 1
    else:
        return n * fact(n-1)

print(fact(5))                  # ⇒ 120
print(fact(5)/fact(2)/fact(3))  # ⇒ 10.0

print(fact(0))  # 0の階乗も問題なく動作 (場合分けの必要はない)
```

- 最後の別解2は，
  $$
  n! = \left\{\begin{array}{ll}1&(n=0 のとき)\\n\times (n-1)!&(それ以外)\end{array}\right.
  $$
  という漸化式を利用している．
- この `fact`関数のように，関数の本体の中から自分自身 (`fact`関数)
  を呼び出す関数は再帰関数と呼ばれる．
- 再帰関数を用いると，さまざまな計算を簡潔に記述できることも多いが，この授業の範囲を超えるので，これ以上は触れないこととする．

### 例題4

- 非負整数の引数 n と k (k≦n) を取り，二項係数 ${}_n{\mathrm C}_{\,k}$ を計算して，その値を戻り値として返す関数 comb(n, k) を定義してみよう．
- 例題3で定義した fact(n) を用いて作ること．
- math.comb は使わないで！
- 戻り値には return 文を用いる．
- 二項係数の計算式は以下の通り（$n!$ は $n$ の階乗を表す）
  $$
  \displaystyle{}_n{\mathrm C}_{\,k} = \frac{n!}{k!\,(n-k)!}
  $$

```python
def comb(n, k):
    c = fact(n)/fact(k)/fact(n-k)
    return c

print(comb(5,2))    # ⇒ 10.0
```

```python
# 別解

def comb(n, k):
    return fact(n)/fact(k)/fact(n-k)

print(comb(5,2))    # ⇒ 10.0
```

```python
# 参考 (fact を使わない再帰関数版)

def comb(n, k):
    if k==0 or k==n:
        return 1
    else:
        return comb(n-1, k-1) + comb(n-1, k) 

print(comb(5,2))    # ⇒ 10
```
