# 「関数の定義」の要約・補足

## 関数の定義

- 関数は以下のように定義する ( `処理1`, `処理2`, ... という内容を，`関数名` という名前で定義する)．

  ```python
  def 関数名(引数名1, 引数名2, ...):
      処理1
      処理2
      ...
  ```

- 定義内容 (本体とも呼ばれる) ( `処理1`，`処理2`, ...) は4字下げる．
- `引数名1`, `引数名2`, ... は，<u>仮引数</u>と呼ばれ，関数を呼び出した際に値が代入されるローカル変数 (後述) のように扱われる．
- 関数を呼び出す際には，関数名の後の括弧の中に，仮引数に代入する値のための式を並べる．これを<u>実引数</u>という．
- 関数は，単独で文として呼び出されたり，式の一部として (値を持った部分式として) 呼び出されたりする．
- 関数の<u>戻り値</u>（関数を式の中で呼び出した時の式としての値）は，<u>retrun</u> 文を用いて与える．

  ```python
  return 値
  ```
  - 関数の本体で retrun 文が実行されると，関数の実行を直ちに中断し，与えられた値を戻り値として，呼び出されたところに戻る．(break と少し似ているかもしれない)
- 値が必要ない関数 (文として呼び出され，式の一部としては呼び出されない関数) の場合，return 文は必要ないが，式の一部として呼び出される関数は，必ず return 文で戻らなければならない．

````python
def hello1(who):
    print("Hello", who)

hello1("everyone")              # 文としての呼び出し．値はない．
hello1("anyone")
````

````python
def hello2(who):
    return "Hello " + who

print(hello2("everyone") + ".") # 式として呼び出し，その戻り値に ".” を付けて出力
print(hello2("everyone"), hello2("anyone"))
````

````python
def hello3(who):
    print("Hello", who)
    return "How are you?"
    print("I'm fine!")   # 上の return で関数の実行が中断されるので，ここは実行されない．

print(hello3("everyone"))
````

## ローカル変数とグローバル変数

- <u>ローカルスコープ</u> - 関数定義の中
  - ここで代入された変数 (<u>ローカル変数</u>)，定義された関数，import されたモジュールは，このローカルスコープ内でのみ使用可能．
  - その関数が呼び出されるごとに変数 (等) が作られ，関数の実行が終了すると消去される．

````python
# 円の半径 (radius) から，円周と面積を表示する関数
#   radius    : 仮引数（ローカル変数）
#   math      : ローカルにインポートされたモジュール
#   enshu     : ローカル変数
#   circleArea: ローカルに定義された関数
def test(radius):
    import math
    enshu = 2 * math.pi * radius    # ローカル変数
    def circleArea(r):
        return math.pi*r*r
    print(enshu)
    print(circleArea(radius))

test(10)                # 正しく機能する

print(enshu)            # エラー
print(math.pi)          # エラー
print(circleArea(5))    # エラー
````

- <u>モジュールスコープ</u> - ファイル内で関数定義以外の部分
  - ここで代入された変数 (<u>グローバル変数</u>)，定義された関数，import されたモジュールは，このファイル内 (モジュール内) のどこでも (関数定義の中でも) 使用できる．
  - ただし，グローバル変数にローカルスコープで代入することはできない．(代入しようとすると，同じ名前のローカル変数が別に作られる － 下の例の `enshu` を見よ)

````python
radius = 10             # グローバル変数
enshu = 0               # グローバル変数
import math             # グローバルにインポートされる

def circleArea(r):      # グローバルに定義される
    return math.pi*r*r

# 円の半径 (radius: グローバル変数) から，円周と面積を表示する関数
#   enshu   : ローカル変数（グローバルな enshu とは別に作られる）
def test():
    enshu = 2 * math.pi * radius    # 新たに enshu というローカル変数が作られる
    print(enshu)
    print(circleArea(radius))

test()                  # 正しく機能する

print(enshu)            # グローバル変数の方の enshu の値が表示される
print(math.pi)          # 正しく機能する
print(circleArea(5))    # 正しく機能する
````

## 関数の呼び出し方

- 引数指定の2つの方法
  - <u>位置引数</u> - 実引数の順序を仮引数の順序に合わせる．
  - <u>キーワード引数</u> - 仮引数の名前を指定して実引数を与える．

````python
def func(arg1, arg2, arg3):
    print("arg1は:", arg1, ", arg2は:", arg2, ", arg3は:", arg3)

func(1, 2, 3)                   # 位置引数
func(arg1=1, arg2=2, arg3=3)    # キーワード引数
func(arg2=2, arg1=1, arg3=3)    # キーワード引数の順序は任意
func(1, arg3=3, arg2=2)         # 混在
````

- デフォルト引数 (引数のデフォルト値) - 呼び出し時に実引数を省略した場合に仮引数に与える値
  - 関数定義の際の仮引数に `引数名=デフォルト値` という形で設定する．

````python
def func(arg1, arg2=22, arg3=33):
    print("arg1は:", arg1, ", arg2は:", arg2, ", arg3は:", arg3)

func(1, 2, 3)   # 省略なし
func(1, 2)      # 第3引数 (arg3) を省略
func(1)         # 第2引数 (arg2) と第3引数 (arg3) を省略
func()          # エラー（デフォルト値を設定していない引数は省略できない）
# func(1, , 3) のように，第2引数 (arg2) だけを省略することはできない
````

````python
def func(arg1, arg2=22, arg3=33):
    print("arg1は:", arg1, ", arg2は:", arg2, ", arg3は:", arg3)

func(1, arg2=2)         # 第3引数 (arg3) を省略
func(arg3=3, arg1=1)    # 第2引数 (arg2) だけを省略
````
