**例題1** (絶対値)

- `input()` で入力した数の絶対値を計算して表示するプログラムを，組込み関数 `abs()` を使わずに書いてみよう．
- 入力した数字列は `float()` を用いて数値に変換する．

```python
x = float(input())

if x < 0:
    absx = -x
else:
    absx = x

print(x, "の絶対値は", absx, "です．")
```

**例題3** (3数の最大値)

- 入力した3つの数の最大値を計算して表示するプログラムを，組込み関数 `max()` を使わずに書いてみよう．
- 入力した数字列は `float()` を用いて数値に変換する．

```python
x = float(input())
y = float(input())
z = float(input())

if x > y:
    if x > z:
        m = x
    else: # x <= z
        m = z
else: # x <= y
    if y > z:
        m = y
    else: # y <= z
        m = z

print(x, y, z, "の最大値は", m, "です．")
```

〔別解〕

```python
x = float(input())
y = float(input())
z = float(input())

m = x       # {x}の最大値 は x
            # この時点では，m は {x}の最大値
if y > m:   # {y}∪{x}={x,y}の最大値 は...
    m = y   #     y と {x}の最大値 の大きい方
            # この時点では，m は {x,y}の最大値
if z > m:   # {z}∪{x,y}={x,y,z}の最大値 は...
    m = z   #     z と {x,y}の最大値 の大きい方
            # この時点では，m は {x,y,z}の最大値

print(x, y, z, "の最大値は", m, "です．")
```

**例題4** (2次方程式の解)

- 二次方程式 $ax^2+bx+c=0$ の係数 $a,\,b,\,c$ から解 $\alpha,\,\beta$ を計算して表示させてみよう．

  $$
  \left\{\begin{array}{rcl}
  \alpha &=& \displaystyle\frac{-b+\sqrt{b^2 - 4ac}}{2a}\\\\
  \beta  &=& \displaystyle\frac{-b-\sqrt{b^2 - 4ac}}{2a}
  \end{array}\right.
  $$

- $a,\,b,\,c$ を入力して，$\alpha,\,\beta$ を表示させる．
- 但し，解がない場合は「解なし」と表示させること．
- なお，入力した数字列は `float()` を用いて数値に変換する．

［ヒント］平方根の計算には，[mathモジュール](https://docs.python.org/ja/3.8/library/math.html) の [`math.sqrt(…)`関数](https://docs.python.org/ja/3.8/library/math.html#math.sqrt) を用いるとよい． 

```python
a = float(input("xの2乗の係数を入力してください．"))
b = float(input("xの係数を入力してください．"))
c = float(input("定数項の値を入力してください．"))

import math

D = b*b - 4*a*c     # 判別式

if D<0:
    print("解なし")
else:
    print((-b+math.sqrt(D)) /2 /a, (-b-math.sqrt(D)) /2 /a)
```

〔別解〕

```python
from math import sqrt

a = float(input("xの2乗の係数(≠0)を入力してください．"))
b = float(input("xの係数を入力してください．"))
c = float(input("定数項の値を入力してください．"))

if a==0:
    print("エラー！")
else:
    print(f"方程式 ({a})x^2 + ({b})x + ({c}) = 0 の解は？")
    D = b*b - 4*a*c     # 判別式

    if D<0:
        print("解なし")
    elif D==0:
        print(-b /2 /a, "（重解）")
    else:
        print((-b+sqrt(D)) /2 /a, (-b-sqrt(D)) /2 /a)
```

**例題5** (おみくじ2)

- 2/10 の確率で「大吉」，4/10 の確率で「吉」，3/10 の確率で「小吉」，1/10 の確率で「凶」を表示するプログラムを書いてみよう．

［ヒント］`random.random()` を使うことを考えてみよ．

```python
import random

rnd = random.random()

if rnd < 0.2:
    print("大吉")
elif rnd < 0.6:     # = 0.2 + 0.4
    print("吉")
elif rnd < 0.9:     # = 0.6 + 0.3
    print("小吉")
else:
    print("凶")
```

〔別解〕

```python
import random

rnd = random.randint(1, 10)

if rnd <= 2:
    print("大吉")
elif rnd <= 6:     # = 2 + 4
    print("吉")
elif rnd <= 9:     # = 6 + 3
    print("小吉")
else:
    print("凶")
```

**例題6** (4数の最大値)

- 入力した4つの数の最大値を計算して表示するプログラムを，組込み関数 `max()` を使わずに書いてみよう．
- 入力した数字列は `float()` を用いて数値に変換する．

```python
x = float(input())
y = float(input())
z = float(input())
w = float(input())

m = x
if y > m:
    m = y
if z > m:
    m = z
if w > m:
    m = w

print(x, y, z, w, "の最大値は", m, "です．")
```
