# 「条件式と分岐」の要約・補足

## プログラムの処理方法

1. 処理を上から順番に一つずつ実行する **順次処理**
2. 条件によって処理を選択する **分岐処理**
3. 条件が満たされている間、処理を繰り返す **反復処理**

## 条件式

- 条件式の値 `True`，`False` は**ブール型**で，各々 真，偽 を表す

### 演算子

| 演算子 | 条件 |
|-------|------|
| a < b | a は b より小さい |
| a <= b | a は b と等しいか小さい |
| a > b | a は b より大きい |
| a >= b | a は b と等しいか大きい |
| a == b | a と b は等しい|
| a != b | a と b は等しくない|

- どの演算子も，文字列の比較にも使える．
  - 文字列の大小比較は，辞書順の比較になる
  - 文字の大小関係は以下の通り：
    $$
    \mathrm{0 < 1 < \cdots < 9 < A < B < \cdots < Z< a < b < \cdots < z}
    $$

````python
print(100 < 20)         # 数値の比較なので False となる
print('100' < '20')     # '1' < '2' なので True となる
````

````python
print('a' < 'B')        # 'a' > 'B' なので False となる
````

````python
print('bcd' < 'a')      # 'b' > 'a' なので False となる
print('bcd' < 'b')      # 'b' == 'b' であるが，辞書順では後続文字のある方が大きいので False となる
print('bcd' < 'c')      # 'b' < 'c' なので True となる
````

### 文字列のメソッド

- `isdecimal()` メソッド： 全て数字なら `True`，それ以外は `False` を返す
- `isalpha()` メソッド： 全てアルファベットなら `True`，それ以外は `False` を返す  
  但し，平仮名や片仮名や漢字もアルファベットとみなされる（全角数字は除く）

````python
print("123は数字ですか？", "123".isdecimal())
print("abcは数字ですか？", "abc".isdecimal())
print("一二三は数字ですか？", "一二三".isdecimal())
print("１２３は数字ですか？", "１２３".isdecimal())
print()
print("123はアルファベットですか？", "123".isalpha())
print("abcはアルファベットですか？", "abc".isalpha())
print("あいうはアルファベットですか？", "あいう".isalpha())
print("アイウはアルファベットですか？", "アイウ".isalpha())
print("一二三はアルファベットですか？", "一二三".isalpha())
print("１２３はアルファベットですか？", "１２３".isalpha())
````

## if 文

- else 節がない（elif 節もない）場合

    ```Python
    if 条件式:
        処理1
        処理2
        ...
    ```

- else 節があり，elif 節がない場合

    ```python
    if 条件式:
        処理1
        処理2
        ...
    else:
        処理3
        処理4
        ...
    ```

- elif 節も else 節もある場合

    ```python
    if 条件式1:
        処理1
        ...
    elif 条件式2:
        処理2
        ...
    elif 条件式3:
        処理3
        ...
    ...
    else:
        処理n
        ...
    ```

- elif 節があり，else 節がない場合

    ```python
    if 条件式1:
        処理1
        ...
    elif 条件式2:
        処理2
        ...
    elif 条件式3:
        処理3
        ...
    ...
    ```

- いずれの場合も `if`, `elif`, `else` の先頭の位置（開始位置）はそろえて，
  `処理` はインデント（字下げ）すること
- `if` あるいは `elif` 直後の `条件式` を順に計算し，
  最初に `True` になった条件の直後の `処理`（インデントされている範囲）のみを実行する
  - 全ての `条件式` が `False` だった場合：  
    `else` 節があれば，その `else` 節の `処理` のみを実行する  
    `else` 節がなければ，何もしない

````python
a = int(input())    # 入力文字列を数値に変換して変数a に代入する
print("a =", a)
if a == 100:        # a が 100 と等しければ、字下げされた処理を実行する
    print("100点満点!")
    print("おめでとう!")
print("バイバイ!")  # if文の次の文なので，常にif文の次に実行される
````

### `random` - 乱数関数を含むモジュール
- `import random` として使用する
- `random.randint(a, b)` ： a 以上 b 以下の整数 (int型) の一様な乱数を返す
- `random.random()` ： 0.0 以上 1.0 未満の実数 (float型) の一様な乱数を返す

その他の関数もあるが，ここでは省略

````python
import random
print(random.randint(1, 10))    # 1 ～ 10 の範囲の一様な整数乱数
print(random.randint(10, 19))   # 10 ～ 19 の範囲の一様な整数乱数
print(random.random())          # 0 以上 1 未満の範囲の一様な実数乱数
````

````python
from random import randint, random
print(randint(1, 10))    # 1 ～ 10 の範囲の一様な整数乱数
print(randint(10, 19))   # 10 ～ 19 の範囲の一様な整数乱数
print(random())          # 0 以上 1 未満の範囲の一様な実数乱数
````
