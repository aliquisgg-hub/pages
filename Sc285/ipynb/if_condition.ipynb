# 条件式と分岐

1. 処理を上から順番に一つずつ実行する **順次処理**
2. 条件によって処理を選択する **分岐処理**
3. 条件が満たされている間、処理を繰り返す **反復処理**

## 文字列の比較
- `==` 演算子は，2つの文字列が同じ値の時 `True` を返す
- `<`, `<=` 等の比較演算子は，辞書順による文字列の大小を比較する

1. 数字は、文字の `0` が最小、`9` が最大(0 < 1 < 2 < ... < 9)
1. アルファベットでは、`a` が最小、`z` が最大 (a < b < c < ... < z)
1. 大文字は小文字より小さい (A < a, B < b, ...)
1. 数字は、アルファベットよりも小さい (0 < 9 < A)

注） $\mathrm{0 < 1 < \cdots < 9 < A < B < \cdots < Z< a < b < \cdots < z}$

````python
'123' < '456'
````

````python
'python' < 'Python'
````

````python
'Python-1' < 'Python-a'
````

## if 文 による条件分岐

```Python
if 条件式:
    処理1
    処理2
    ...
```

````python
a = 100 # 変数a に 100を設定する

if a == 100:  # a が 100 と等しければ、print()関数を実行する
    print("100点満点!")
````

### if 文の書き方
1. まず `if 条件式:` という行で始める（`:` を忘れないように！）
1. その次の行から，`条件式` が `True` の場合に実行する処理をインデントして記述する  
   **インデント**： 行の先頭にスペースを4文字入れて，命令の始まりを下げる

   ```python
   if a == 100:
      処理1
      処理2
      処理3
   ^^^^
   スペースを4文字入力
   ```
1. 条件式が `True` なら，字下げされた処理を順に実行する．  
   条件式が `False` なら，何も実行せずに（字下げされた処理の）次へ進む

````python
a = 99 # 変数a に 99 を設定する

if a == 100:  # a が 100 と等しければ、print()関数を実行する
    print("100点満点!")
````

## else 節
条件が**満たされなかった**ときだけ実行する処理を指定する

```python
if 条件式:
    処理1
    処理2
    ...
else:
    処理3
    処理4
    ...
```

- `条件式` の結果が `True` になるときは，`if:` に続く処理1，処理2，... が実行される
- `条件式` の結果が `False` のときは，`else:` に続く処理3、処理4、... が実行される

`else:` 以降の部分を **else節** と言う．

注） else節もインデントを忘れないように

````python
a = 100 # 変数aに 100 を設定する

if a == 100:  # a が 100 と等しければ、print()関数を実行する
    print("100点満点!")
else:
    print("失格!")
````

````python
a = 0 # 変数a に 0 を設定する

if a == 100:  # a が 100 と等しければ、print()関数を実行する
    print("100点満点!")
else:
    print("失格!")
````

## 比較以外の条件式

`isdecimal()` メソッド： 全て数字かどうか調べる

````python
print("123は数字ですか？", "123".isdecimal())
print("abcは数字ですか？", "abc".isdecimal())
````

````python
string = input("文字列を入力してください:")
if string.isdecimal():
    print(string, "は数字です")
else:
    print(string, "は数字ではありません")
````

## elif 節

`isalpha()` メソッド： 全てアルファベットかどうか調べる  
※ 平仮名や片仮名や漢字もアルファベットとみなされる

````python
print("123はアルファベットですか？", "123".isalpha())
print("abcはアルファベットですか？", "abc".isalpha())
````


**elif 節**

`if` で指定した条件が満たされなかった時に検査する条件を指定する

```python
if 条件式1:
    処理1
    ...
elif 条件式2:
    処理2
    ...
elif 条件式3:
    処理3
    ...
else:
    処理n
    ...
```

`elif ～` で追加された条件と処理の部分を **elif節** と言う．

- elif 節の処理の部分もインデントが必要
- `条件式1`, `条件式2`, ... を順に判定し，最初に `True` になった条件の直後の処理（のみ）を実行する
- 全ての条件が満たされなかった（`False` になった）ときは，else 節の処理（のみ）を実行する
  - 全ての条件が満たされず，else 節もないときは，何もしないで次に進む

````python
string = input("文字列を入力してください:")
if string.isdecimal():
    print(string, "は数字です")
elif string.isalpha():
    print(string, "はアルファベットです")
else:
    print(string, "は数字でもアルファベットでもありません")
````
