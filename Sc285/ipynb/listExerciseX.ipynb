### 例題1
- 数値のリスト l を引数として取り，そのリスト中の最大値を戻り値として返す関数 maximum(l) を作成してみよ．
  ただし，組み込み関数は使用しないこと．

ヒント：
- それまでに見つかった最大値を保存しておく変数を用意し，リストの要素と順に比較して，大きい方を残していけばよい．

```python
def maximum(l):
    maxi = l[0]
    for x in l:
        if x > maxi:
            maxi = x
    return maxi

nums = [1, -4, 2, -8, 5, -7]
print(maximum(nums))    # ⇒ 5
```

### 例題2
- 引数としてリスト l と2つの添え字 i, j を取り，リストlの i 番の要素と j 番の要素を入れ替える関数 swap(l, i, j) を作成してみよ．

ヒント：

- 2つの値を入れ替えるためには，値を一時的に退避する変数を用意するとやりやすいかも．
- 複数の変数に同時に代入する機能（フィボナッチ数列で使ったやつ）を用いれば，退避用の変数はいらないかも．

※ 戻り値は不要なので，return 文はいらない．

```python
def swap(l, i, j):
    z = l[j]    # l[j] の退避
    l[j] = l[i]
    l[i] = z    # l[j] を戻す

nums = [1, -4, 2, -8, 5, -7]
swap(nums, 0, 1)    # nums の順序が入れ替わる
print(nums)         # ⇒ [-4, 1, 2, -8, 5, -7]
swap(nums, 1, 3)    # nums の順序が入れ替わる
print(nums)         # ⇒ [-4, -8, 2, 1, 5, -7]
```

```python
def swap(l, i, j):
    l[i], l[j] = l[j], l[i]

nums = [1, -4, 2, -8, 5, -7]
swap(nums, 0, 1)    # nums の順序が入れ替わる
print(nums)         # ⇒ [-4, 1, 2, -8, 5, -7]
swap(nums, 1, 3)    # nums の順序が入れ替わる
print(nums)         # ⇒ [-4, -8, 2, 1, 5, -7]
```

### 例題3
- 数値のリスト l を引数として取り，その各要素を2乗したもののリストを生成する関数 sqList1(l) を作成してみよ．ただし，
  - 生成するリストは，元のリストに上書きすること
  - 戻り値は不要

ヒント：
- 各 l[i] に，l[i]の2乗を代入すればよい．
-　リストに上書きするには添え字(番号)が必要なので，for ではなく while を用いる．

```python
def sqList1(l):
    i=0
    while i<len(l):     # len(l) は、l の長さ
        l[i] = l[i]*l[i]
        i+=i+1

list1 = [1, -4, 2, -8, 5, -7]
sqList1(list1)       # list1 の内容が書き替わる
print(list1)         # ⇒ [1, 16, 4, 64, 25, 49]
sqList1(list1)       # list1 の内容が書き替わる
print(list1)         # ⇒ [1, 256, 16, 4096, 625, 2401]
```

```python
def sqList1(l):
    for i in range(len(l)): # len(l) は、l の長さ
        l[i] = l[i]*l[i]

list1 = [1, -4, 2, -8, 5, -7]
sqList1(list1)       # list1 の内容が書き替わる
print(list1)         # ⇒ [1, 16, 4, 64, 25, 49]
sqList1(list1)       # list1 の内容が書き替わる
print(list1)         # ⇒ [1, 256, 16, 4096, 625, 2401]
```

### 例題4
- 数値のリスト l を引数として取り，その各要素を2乗したもののリストを生成する関数 sqList2(l) を作成してみよ．ただし，
  - 生成するリストは，元のリストとは別に新たに生成すること．
  - 生成したリストは戻り値として値を返すこと（return 文を用いる）

ヒント：
- 空リスト (`[]`) を生成し，l の各要素の2乗を順に末尾に追加していけばよい． 

```python
def sqList2(l):
    m=[]
    for x in l:
        m.append(x*x)
    return m

nums = [1, -4, 2, -8, 5, -7]
print(sqList2(nums))            # ⇒ [1, 16, 4, 64, 25, 49]
print(sqList2(sqList2(nums)))   # ⇒ [1, 256, 16, 4096, 625, 2401]
print(nums)                     # ⇒ [1, -4, 2, -8, 5, -7]           # nums は不変
```

```python
def sqList2(l):
    m = l[:]    # l のコピーを m に作る 
    i=0
    while i<len(l):
        m[i]=l[i]*l[i]
        i=i+1
    return m

nums = [1, -4, 2, -8, 5, -7]
print(sqList2(nums))            # ⇒ [1, 16, 4, 64, 25, 49]
print(sqList2(sqList2(nums)))   # ⇒ [1, 256, 16, 4096, 625, 2401]
print(nums)                     # ⇒ [1, -4, 2, -8, 5, -7]           # nums は不変
```

### 例題5
- 数値のリスト l を引数として取り，その各要素のうち，正の数のみからなるリストを生成する関数 posiList(l) を作成してみよ．ただし，
  - 生成するリストは，元のリストとは別に新たに生成すること．
  - 生成したリストは戻り値として値を返すこと（return 文を用いる）

ヒント：
- 空リスト (`[]`) を生成し，l の各要素のうち，条件を満たしているもののみを順に末尾に追加していけばよい． 

```python
def posiList(l):
    pl = []                         # pl に空リストを用意して
    for x in l:                     # l の各要素 x について
        if x > 0:                   # x が 0より大きいときのみ
            pl.insert(len(pl), x)   # pl の末尾に x を追加する
    return pl                       # 戻り値は pl

nums = [1, -4, 2, -8, 5, -7]
print(posiList(nums))           # ⇒ [1, 2, 5]
print(nums)                     # ⇒ [1, -4, 2, -8, 5, -7]           # nums は不変
```

- [append](https://docs.python.org/ja/3.8/tutorial/introduction.html?highlight=list%20append#lists) メソッドを使えば，より簡潔に書ける．

```python
def posiList(l):
    pl = []
    for x in l:
        if x > 0:
            pl.append(x)
    return pl

nums = [1, -4, 2, -8, 5, -7]
print(posiList(nums))           # ⇒ [1, 2, 5]
print(nums)                     # ⇒ [1, -4, 2, -8, 5, -7]           # nums は不変
```

- こんな書き方も ……

```python
def posiList(l):
    return [x for x in l if x > 0]

nums = [1, -4, 2, -8, 5, -7]
print(posiList(nums))           # ⇒ [1, 2, 5]
print(nums)                     # ⇒ [1, -4, 2, -8, 5, -7]           # nums は不変
```

### 例題6
- 数値のリスト l を引数として取り，そのリスト中の最大値<u>の添え字の値</u>を戻り値として返す関数 indexOfMax(l) を作成してみよ．

ヒント：
- それまでに見つかった最大値の添え字の値を保存しておく変数を用意し，リストの要素と順に比較して，大きい方の要素の添え字の値を残していけばよい．
- リストの要素の値だけではなく，添え字(番号)も必要なので，for ではなく while を用いる．


```python
def indexOfMax(l):
    maxv = l[0]             # 最初は (暫定的に) l[0] が最大と仮定し
    maxi = 0                # その添え字を maxi に保存する
    i = 1                   # i を 1 から
    while i < len(l):       # … len(l)-1 まで変えながら反復
        if l[i] > maxv:     # もし l[i] が maxv（それまでの暫定的な最大値）より大きいとき
            maxv = l[i]     # l[i] を (暫定的な) 最大値として
            maxi = i        # その添え字を i を maxi に保存する
        i = i + 1
    return maxi             # 戻り値は maxi

nums = [1, -4, 2, -8, 5, -7]
print(indexOfMax(nums))     # ⇒ 4
```

```python
def indexOfMax(l):
    maxi = 0                # 最初は (暫定的に) l[0] が最大と仮定する
    i = 1                   # i を 1 から
    while i < len(l):       # … len(l)-1 まで変えながら反復
        if l[i] > l[maxi]:  # もし l[i] が l[maxi]（それまでの暫定的な最大値）より大きいとき
            maxi = i        # l[i] を (暫定的な) 最大値として maxi を iの値 で置き換える
        i = i + 1
    return maxi             # 戻り値は maxi

nums = [1, -4, 2, -8, 5, -7]
print(indexOfMax(nums))     # ⇒ 4
```

- [enumerate](https://docs.python.org/ja/3.8/tutorial/datastructures.html#tut-loopidioms) メソッドを使うと ……

```python
def indexOfMax(l):
    maxi = 0                # 最初は (暫定的に) l[0] が最大と仮定する
    for i,x in enumerate(l[1:]):
        if x > l[maxi]:
            maxi = i
    return maxi

nums = [1, -4, 2, -8, 5, -7]
print(indexOfMax(nums))     # ⇒ 4
```

### 例題7
- 数値のリスト l, m を引数として取り，その l と m を連結したリストを生成する関数 concat1(l, m) を作成してみよ．ただし，
  - リスト l の末尾に m の要素を追加する形で実装する．
  - その連結されたリストを戻り値とする．

```python
def concat1(l, m):
    l[len(l):] = m    # l の末尾の部分空リストを m で置き換える
    return l

x = [1, 2, 3]
y = ["A", "B", "C"]
xy = concat1(x, y)

print(x, y, xy)     # y は不変だが，x, xy は共に連結されたリストになる．
```

### 例題8
- 数値のリスト l, m を引数として取り，その l と m を連結したリストを生成する関数 concat2(l, m) を作成してみよ．ただし，
  - 生成するリストは，元のリストとは別に新たに生成すること．
  - 生成したリストを戻り値とする（return 文を用いる）．
  - concat1 を使ってもよい．

ヒント：
- l のコピー(仮に l1 とする)を作って，その l1 と m とで concat1 を呼び出せばよい

```python
def concat2(l, m):
    l1 = l[:]               # l のコピーを作る
    return concat1(l1, m)

x = [1, 2, 3]
y = ["A", "B", "C"]
xy = concat2(x, y)

print(x, y, xy)     # x も y も不変だが，xy は連結されたリストになる．
```
