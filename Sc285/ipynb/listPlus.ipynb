## Pythonの型とオブジェクト

- **オブジェクト**： データ（名前が付けられる）
- **型**： オブジェクトの種類
  |　型     |意味|説明|
  |---------|---|---|
  | int     | 整数 | 整数値を表す値．|
  | float   | 実数 | 実数値を表す値．|
  | str     | 文字列 | 文字の並び．文字のシーケンスともみなせる．|
  | bool    | ブール値 | 真偽を表す値．|
  | module  | モジュール | インポートされたモジュールを表すオブジェクト．<br>さまざまな関数（メソッド）や値（プロパティ）などを内部に含む．|
  | function| 関数 | 関数を表すオブジェクト． |

  ※ 以下については，詳細は後述．

  |　型     |意味|説明|
  |---------|---|---|
  | list    | リスト | データの並びを表すオブジェクト．個々の要素のシーケンスとみなされる．<br>可変（後からデータ要素の変更・追加・削除が可能）．|

  | dict    | 辞書 | キー（見出し）と値のペアの集まり．<br>キーを用いて値をアクセスするコレクションとみなされる．|

## リスト
### リストの生成

- 要素をカンマ `,` で区切って並べ，角括弧 `[]` で括る．

※ リストの長さ（要素数）は，len関数で調べられる

```python
list1 = [1, 4, 2, 8, 5, 7]  # 6個の要素を持つリストが変数 list1 に代入される．
print(list1, len(list1))
```

### 要素の参照  
- リストを表す式（リストが代入されている変数など）の後に，インデックス（添字）を角括弧 `[]` で括って添える．
  - インデックス（添字）は整数値で，要素の順番（先頭の要素を 0 とする）を表す．
  - 文字列と同様に，負数のインデックスも使用できる（末尾の要素を -1 とする）．
  - 文字列と同様に，リストの部分切り出しもできる．

<div style="margin-left: 4em;">



</div>

```python
list1 = [1, 4, 2, 8, 5, 7]

print(list1[0], list1[2], list1[4])         # インデックスによる要素の参照
i=2
print(list1[0*i], list1[1*i], list1[2*i])   # インデックスは数式でもよい
print(list1[0*i] + list1[1*i] + list1[2*i]) # 参照した値は計算にも使える
print(list1[-1], list1[-3], list1[-5])      # インデックスは負数でもよい
```

```python
list1 = [1, 4, 2, 8, 5, 7]

print(list1[1:-1], list1[:3], list1[3:])    # 部分切り出しも可能
print(list1[0:0], len(list1[0:0]))          # list1[0:0] は，空の(要素のない)リスト
print(list1[0:1], len(list1[0:1]))          # list1[0:1] は，要素が1つのリスト(先頭の要素そのものではない)

list2 = list1[1:-1]     # 切り出した部分もリスト

print(list2)
print(list2[0], list2[-1])
```

## リストの操作
### 要素の置き換え(代入)  
- インデックスを使った要素の参照に，新たな値を代入する．

```python
list1 = [1, 4, 2, 8, 5, 7]

list1[0] = "0番"
list1[-1] = "-1番"
print(list1)
```

- 部分切り出しに別のリストを代入することで，その部分だけ別のリストで置き換えることもできる

```python
list1 = [1, 4, 2, 8, 5, 7]

list1[1:-1] = ["中4つの", "置き換え"]
print(list1)

list1[0:1] = [9999]
print(list1)

list1[0:1] = 9999     # これはエラーになる
```

### 要素の挿入  
- insertメソッドを用いる

```python
list1 = [1, 4, 2, 8, 5, 7]
print(list1)

list1.insert(3, "|")            # 真ん中への挿入
print(list1)

list1.insert(0, "(")            # 先頭への挿入
print(list1)

list1.insert(len(list1), ")")   # 末尾への挿入
print(list1)
```

- 部分リストの切り出しを利用してもできる

```python
list1 = [1, 4, 2, 8, 5, 7]
print(list1)

list1[3:3] = ["|"]                      # 真ん中への挿入
print(list1)

list1[0:0] = ["("]                      # 先頭への挿入
print(list1)

list1[len(list1):len(list1)] = [")"]    # 末尾への挿入
print(list1)
```

### 要素の削除  
- del文を用いる

```python
list1 = ["(", 1, 4, 2, "|", 8, 5, 7, ")"]

del list1[-1]   # 末尾の削除
print(list1)

del list1[0]    # 先頭の削除
print(list1)

del list1[3]    # 真ん中の削除
print(list1)
```

- 部分リストの切り出しを利用してもできる

```python
list1 = ["(", 1, 4, 2, "|", 8, 5, 7, ")"]

list1[-1:] = []     # 末尾の削除
print(list1)

list1[:1] = []      # 先頭の削除
print(list1)

list1[3:4] = []     # 真ん中の削除
print(list1)
```

## リストを引数とする関数

- 関数の実引数としてリストを与えることもできる．
- その場合，関数の内部では仮引数をリストとして，通常のリストと同様に扱える．
- ただし，リストに対する要素の置き換え(代入)，要素の挿入，要素の削除は，リスト内部を直接書き換えるものなので，
  実引数として与えられたリストの方でも，内部が書き換えられることになるので，
  (スコープ外の)グローバル変数の値にも影響が及ぶ．（これを<u>副作用</u>という）

```python
def absList(nums):  # 数値のリストを引数 nums として取り，各要素をその絶対値に書き換える関数
    i = 0
    while i < len(nums):
        nums[i] = abs(nums[i])  # nums に代入されているリストを書き換えることになる
        i = i + 1

list1 = [1, -4, 2, -8, 5, -7]
print("absList実行前の list1 =", list1)
absList(list1)                          # list1 に代入されているリストを実引数として関数 absList に渡す
print("absList実行後の list1 =", list1) # 副作用の結果，list1 に代入されているリストも書き換えられている
```
