**例題1** (3数の最大値)

以下のコードの2か所の条件式 (`None` と書かれているところ) を書き換えて，プログラムを完成させよう．  
(`None` は値がないことを表す特殊な定数)

```python
x = float(input())
y = float(input())
z = float(input())

if None:        # None を「x が y や z よりも大きい (小さくない)」ことを表す条件式にする
    m = x
elif None:      # None を「y が z よりも大きい (小さくない)」ことを表す条件式にする
    m = y
else:
    m = z

print(x, y, z, "の最大値は", m, "です．")
```
