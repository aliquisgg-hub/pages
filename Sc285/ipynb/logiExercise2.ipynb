**例題1** (3数の最大値)

以下のコードの2か所の条件式 (`None` と書かれているところ) を書き換えて，プログラムを完成させよう．  
(`None` は値がないことを表す特殊な定数)

```python
x = float(input())
y = float(input())
z = float(input())

if x > y and x > z:     # 「x が y や z よりも大きい」，y < x > z でも OK
    m = x
elif y > z:             # 「y が z よりも大きい」
    m = y
else:
    m = z

print(x, y, z, "の最大値は", m, "です．")
```

**例題2** (SSPR登録チェック)

`phone`, `email`, `appli` の3つの変数のうちの2つ以上が True の場合は **OK**，
それ以外の場合は **NG** を表示するコードを書いてみよう．

- 結果表示には `print()` を使うこと

［ヒント］
- `phone`, `email`, `appli` の3つの変数から2つ選ぶ選び方は3通りあるが，
  その選んだ変数の両方が True である条件を表す条件式を選び方ごとに作って，
  それらをどう使えばよいか考えてみよう．
- あるいは，`phone`, `email`, `appli` の3つの変数のうちの True の数を数える方法を考えてもよい．
- それ以外の方法を考えてもよい．

```python
# 以下の3つの変数の値は，いろいろ変更して試してみること
phone = False
email = True
appli = False

if phone and email:
    print("OK")
elif phone and appli:
    print("OK")
elif email and appli:
    print("OK")
else:
    print("NG")
```

```python
# 以下の3つの変数の値は，いろいろ変更して試してみること
phone = True
email = True
appli = False

if phone and (email or appli):
    print("OK")
elif email and appli:
    print("OK")
else:
    print("NG")
```

```python
# 以下の3つの変数の値は，いろいろ変更して試してみること
phone = True
email = True
appli = False

if (phone and email) or (phone and appli) or (email and appli):
    print("OK")
else:
    print("NG")
```

```python
# 以下の3つの変数の値は，いろいろ変更して試してみること
phone = True
email = True
appli = False

count = 0
if phone:
    count = count + 1
if email:
    count = count + 1
if appli:
    count = count + 1

if count >= 2:
    print("OK")
else:
    print("NG")
```

**補足**

- if節 や elif節 や else節 の処理部分の命令が1つなら，`:` の後に続けて書いてもよい
- `A = A + 1` は，`A += 1` のようにも書ける

```python
# 以下の3つの変数の値は，いろいろ変更して試してみること
phone = True
email = True
appli = False

count = 0
if phone: count += 1
if email: count += 1
if appli: count += 1

if count >= 2:
    print("OK")
else:
    print("NG")
```

**補足**

- 整数の 1 は True，0 は False とみなされるので，
`True`，`False` の代わりに `1`，`0` を使ってもよい．

```python
# True が整数の 1 と等価であり，False が整数の 0 と等価であることを利用する

# 以下の3つの変数の値は，いろいろ変更して試してみること
phone = False
email = True
appli = False

if phone + email + appli >= 2:
    print("OK")
else:
    print("NG")
```
