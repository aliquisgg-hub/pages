# ブール型と論理演算子のまとめ

## ブール型

- **ブール型**は True か False の2つの値からなる型．True と False は，それぞれ <u>真</u> と <u>偽</u> を意味する．
- プログラム中に書かれた `True`, `False` は，それぞれ True, False の値を表す定数となる．

```python
print("コード中に書かれた「.5」は，数値", .5, "を表す定数です．")
print("コード中に書かれた「5.」は，数値", 5., "を表す定数です．")
print("コード中に書かれた「'abc'」は，文字列", 'abc', "を表す定数です．")
print('コード中に書かれた「"abc"」は，文字列', "abc", 'を表す定数です．')
print("コード中に書かれた「True」は，ブール値", True, "を表す定数です．")
print("コード中に書かれた「False」は，ブール値", False, "を表す定数です．")
```

## 論理演算子

ブール型の値に対する演算子  
`A,B`： ブール型の値を持つ式とする
- 二項演算子： `A 演算子 B` の形の式を構成する
  - and 演算子： `A and B`　―　「AかつB」(論理積)
    - `A`, `B` の両方の値が True のときのみ，`A and B` の値は True になる
    - `A`, `B` のどちらか一方の値が False のときは，`A and B` の値は False になる
  - or 演算子： `A or B`　―　「AまたはB」(論理和)
    - `A`, `B` のどちらか一方の値が True のときは，`A or B` の値は True になる
    - `A`, `B` の両方の値が False のときのみ，`A or B` の値は False になる

| `A` | `B` |`A and B`|`A or B`|
|:---:|:---:|:-------:|:------:|
|False|False|False    |False   |
|False|True |False    |True    |
|True |False|False    |True    |
|True |True |True     |True    |

- 単項演算子： `演算子 A` の形の式を構成する
  - not 演算子： `not A`　―　「Aでない」(否定，論理否定)
    - `A` の値が False のとき，`not A` の値は True になる
    - `A` の値が True のとき，`not A` の値は False になる

| `A` |`not A`|
|:---:|:-----:|
|False|True   |
|True |False  |


```python
x = 15
print(18 <= x and x < 75)
x = 20
print(18 <= x and x < 75)
x = 75
print(18 <= x and x < 75)

print()

x = 15
print(x < 18 or 75 <= x)
x = 20
print(x < 18 or 75 <= x)
x = 75
print(x < 18 or 75 <= x)
```

比較演算子については，例えば `A < B and B < C` のような式は，`A < B < C` のように書くこともできる．

```python
x = 15
print(18 <= x < 75)
x = 20
print(18 <= x < 75)
x = 75
print(18 <= x < 75)
print(18 < x > 20)      # 18 < x and x > 20  と同じ
```

以下のような等価な式を知っていると，条件式を簡単にするのに役立つかもしれない．  
（表の **式1** と **式2**，**式3** と **式4** は各々等価．  
　表中の A, B, C はブール型の値を取る任意の式）

|式1|式2| |式3|式4|

| A == True | A | | A == False | not A |
| A != True | not A | | A != False | A |
| not (A == B) | A != B | | not (A != B) | A == B |
| not (A < B) | A >= B | | not (A >= B) | A < B |
| not (A > B) | A <= B | | not (A <= B) | A > B |
| (not A) and (not B) | not (A or B) | | (not A) or (not B) | not (A and B) |
| (A and B) or (A and C) | A and (B or C) | | (A or B) and (A or C) | A or (B and C) |
| (A and C) or (B and C) | (A or B) and C | | (A or C) and (B or C) | (A and B) or C |
| not (not A)) | A | | | |




## ブール値と整数
- `True` は整数の `1`，`False` は整数の `0` と等価とみなされることがある．

```python
print(True,      "のタイプは", type(True))
print(True * 1,  "のタイプは", type(True * 1))
print(False,     "のタイプは", type(False))
print(False * 1, "のタイプは", type(False * 1))

print(True,  "+", True,  "は", True + True)
print(False, "+", False, "は", False + False)
```
