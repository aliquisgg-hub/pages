# ブール型と論理演算子

## ブール型

- ブール型 は `True` か `False` どちらかの値をとり，`True` と `False` は，それぞれ 真 と 偽 を意味する．

````python
print(1 > 2)
````

````python
print("abc" == "abc")
````

- プログラム中に書かれた `True`, `False` は，それぞれ `True`, `False` の値を表す定数となる．

````python
true_value = True
print("true_value は", true_value)

false_value = False
print("false_value は", false_value)
````

## 論理演算子

- 複数の条件を組み合わせて条件を判定する演算子を，**論理演算子** (または**ブール演算子**) と言う．
- ブール型の値に対する演算子と考えられる．
- 以下の3種類がある．
  - and 演算子
  - or 演算子
  - not 演算子

## and 演算子

- 「10歳以上で身長は120cm以上の方のみお乗りいただけます」

  1. 年齢 `age` が 10 またはそれより大きい
  2. 身長 `height` が 120 またはそれより大きい

  の両方を満たせば，乗ってもよい．

````python
age = 11        # 例として、11歳 身長130cmとする
height = 130

if (10 <= age) and (120 <= height): # True and True と同じ
    print("お乗りいただけます")
else:
    print("ご遠慮ください")
````

- **`and` 演算子は，左右に指定された値の両方が `True` の場合，`True` を返す**（それ以外は `False` を返す）．

````python
age = 11        # 例として、11歳 身長110cmとする
height = 110    # この行を 130 -> 110に変更

if (10 <= age) and (120 <= height): # True and False と同じ
    print("お乗りいただけます")
else:
    print("ご遠慮ください")
````

### 論理積

```python
C = A and B
```
- 変数 `C` の値は，`A` と `B` の値によって次のようになる．

|  C  | |  A  |  B  |
|:---:|-|:---:|:---:|
|False| |False|False|
|False| |False|True |
|False| |True |False|
|True | |True |True |

- `A` と `B` の両方が `True` のときのみ，結果 `C` の値が `True` となる．
- `and` のように，すべての入力値が `True` のときのみ結果が `True` となる論理演算は，**論理積** ともいう．



## or 演算子

- 「10歳以下の方と、80歳以上の方はご遠慮ください」

  1. 年齢 `age` が10以下である
  1. 年齢 `age` が80以上である

  ２つの条件の，どちらかが満たされていれば搭乗禁止．

````python
age = 90    # 年齢を90歳とする

if (age <= 10) or (80 <= age):  # False or True と同じ
    print("ご遠慮ください")
else:
    print("お乗りいただけます")
````

- **`or` 演算子は，左右に指定された値のどちらか一方，または両方が `True` の場合，`True` を返す**（両方が `False` なら `False` を返す）．

````python
age = 50    # 年齢を50歳とする

if (age <= 10) or (80 <= age):  # False or False と同じ
    print("ご遠慮ください")
else:
    print("お乗りいただけます")
````

### 論理和

```python
C = A or B
```
- 変数 `C` の値は，`A` と `B` の値によって次のようになる．

|  C  | |  A  |  B  |
|:---:|-|:---:|:---:|
|False| |False|False|
|True | |False|True |
|True | |True |False|
|True | |True |True |

- `A` と `B` の両方が `False` のときのみ，結果 `C` の値が `False` となる．
- `or` のように，すべての入力値が `False` のときのみ結果が `False` となる論理演算は，**論理和** ともいう．


## not 演算子
- `not` 演算子は，指定した式の値を**反転**し，値が True(真) なら False(偽)を返し，False(偽)ならTrue(真) を返す．
- 次のように記述する．

  ```python
  not 式
  ```
- 「10歳未満**以外**はお乗りいただけます」

  1. 年齢 `age` が10未満である

  この条件を**満たしていない**ならば，乗ってもよい

````python
age = 20    # 年齢は20歳

if not (age < 10):      # not False と同じ
    print("お乗りいただけます。")
````

### 否定 (論理否定)

```python
B = not A
```
- 変数 `C` の値は，`A` と `B` の値によって次のようになる．

|  B  | |  A  |
|:---:|-|:---:|
|True | |False|
|False| |True |

- `A` が `False` ならば `B` は `True`，`A` が `True` ならば `B` は `False` となる．
- `not` のように，入力値が反転した値が結果となる論理演算を，**否定 (論理否定)** ともいう．
