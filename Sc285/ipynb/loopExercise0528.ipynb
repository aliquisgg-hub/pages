# 「while文によるループ」の例題

### 例題1：

- 以下のような出力を得るコードを，while文を使って書いてみよう
- なお，`print()` は 1か所しか使わないようにすること

<hr>

```
1
2
3
4
5
```
<hr>

［ヒント］loopPlus.ipynb の 例5 あたりを参考に ...

````python
i = 1
while i < 6:
    print(i)
    i = i + 1
````

````python
i = 0
while i < 5:
    print(i+1)
    i = i + 1
````

````python
i = 0
while i < 5:
    i = i + 1
    print(i)
````

［ヒント］1ずつ足す代わりに，1ずつ引いてみれば ...

### 例題5：

- 複数個の整数を順に入力して，合計を計算して出力するコードを，while を使って書いてみよう
- なお，個数はあらかじめ決めず，0 が入力されるまでの合計を求めるものとする
- 入力は `int()` で整数値に変換するのを忘れないように

［ヒント］
- 例題4 と同様に，適当な変数 (仮に `sum` とする) を 0
  にしておき，入力した整数をそこに繰り返し足していくことを考えよう
- loopPlus.ipynb の 例6 (下のコード) を参考にして …
  - 0 の入力で終了することにして …
  - 文字列を連結する代わりに，数値を足せば …
- 変数は分かりやすい名前にしておこう

````python
# 例6
A = ""
text = input()  
while text != "finish":
    A = A + text + ", "
    text = input()  
print(A)
````
