# 「while文によるループ」の例題

### 例題1：

- 以下のような出力を得るコードを，while文を使って書いてみよう
- なお，`print()` は 1か所しか使わないようにすること

<hr>

```
1
2
3
4
5
```
<hr>

［ヒント］loopPlus.ipynb の 例5 あたりを参考に ...

````python
# 解答例1
A = 1
while A < 6:
    print(A)
    A = A + 1
````

````python
# 解答例2
A = 1
while A <= 5 :
    print(A)
    A = A + 1
````

````python
# 解答例3
A = 0
while A < 5 :
    A = A + 1
    print(A)
````

### 例題3：

- 以下のような出力を得るコードを，while文を使って書いてみよう

<hr>

```
1, 2, 3, 4, 5, 
```
<hr>

［ヒント］loopPlus.ipynb の 例3 を参考にして ...

````python
# 解答例1
A = 1
B = str(A) + ", "
while A < 5:
    A = A + 1
    B = B + str(A) + ", "

print(B)
````

````python
# 解答例2
A = 0
B = ""
while A < 5:
    A = A + 1
    B = B + str(A) + ", "

print(B)
````

````python
# 解答例3
B = ""
A = 1
while A <= 5:
    B = B + str(A) + ", "
    A = A + 1

print(B)
````

### 例題4：

- 複数個の整数を順に入力して，合計を計算して出力するコードを，while を使って書いてみよう
- なお，個数はあらかじめ変数 `n` で決めておくものとします
- 入力は `int()` で整数値に変換するのを忘れないように

［ヒント］
- 初めに適当な変数 (仮に `sum` とします) を 0 にしておく
- 以下を `n` 回繰り返す
  - 整数を入力
  - 入力した整数を `sum` に足す
- `sum` を表示する


````python
# 解答例1
n = 5   # 5個の数値を合計する場合
sum = 0
A = 0

while A < n:
    X = int(input("Enter Number"))
    print(X)    # 動作確認用
    sum = sum + X
    A = A + 1

print(sum)
````

````python
# 解答例2
n = 5
sum = 0

while n>0:
    sum += int(input("Enter Number"))
    n -= 1

print(sum)
````

````python
# 解答例1
n = 45      # 45 は一例
i = 0
buf = ""

while i < n:
    i = i + 1
    buf = buf + str(i) + ", "
    if i % 10 != 0:     # この2行が
        continue        # 追加分
    print(buf)
    buf = ""
print(buf)
````

````python
# 解答例2
n = 45      # 45 は一例
i = 0
buf = ""

while i < n:
    i = i + 1
    buf = buf + str(i) + ", "
    if i % 10 == 0:     # 以下3行が
        print(buf)      # 修正されている
        buf = ""        # 
print(buf)
````

````python
# 解答例1
max = int(input())
print("-", max)    # 動作確認用
while True:
    x = input()
    print("-", x, max)    # 動作確認用
    if x == "":
        break
    elif int(x) > max:
        max = int(x)
print(max)
````

````python
# 解答例2
mx = -float("inf")  # マイナス無限大(どんな数よりも小さい)
while True:
    x = input()
    print("-", x, mx)    # 動作確認用
    if x == "":
        break
    elif int(x)>mx:
        mx = int(x)
print(mx)
````

````python
# 解答例1
n = 20  # 仮の値

F0 = 0
print(F0)
F1 = 1
print(F1)
while n > 2:
    F1, F0 = F1 + F0, F1
    print(F1)
    n = n - 1
````

````python
# 解答例2
n = 20  # 仮の値

F0 = 0
print("F0 =", F0)
F1 = 1
print("F1 =", F1)

i = 2
while i < n:
    F1, F0 = F1 + F0, F1
    print("F"+str(i), "=", F1)
    i = i + 1
````

````python
# 解答例3
n = 30  # 仮の値

F1, F0 = 1, 0
buf = "0, 1, "

i = 2
while i < n:
    i = i + 1
    F1, F0 = F1 + F0, F1
    buf = buf + str(F1) + ", "
    if i % 10 != 0:
        continue
    print(buf)
    buf = ""
print(buf)
````
