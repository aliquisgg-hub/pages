# 「while文によるループ」の要約・補足

## 代入文と等号

- <span style="font-size:150%;">代入文は等式<u>ではない</u> ― これ大事！</span>
  - 右辺を計算した結果を左辺の変数に*代入*する働き (その変数の新しい値とする) を持つ命令である．
  - だから，左辺は必ず変数 (の働きを持つもの) である．
  - 両辺が<u>等しい</u>ということを表すわけ<u>ではない</u>．  
    &nbsp;
- 例えば，`A` に数値が代入されている状態では，`A = A + 1` は，`A` の値に 1 を足した値を `A` の新しい値とする働きを持つ．  
  すなわち，`A` の値を 1 増やすことになる．例えば，`A` の値が 3 ならば，`A = A + 1` を実行した後の `A` の値は 4 になる．
  - 通常の等式と考えると，両辺から `A` を引けば，`0 = 1` となって矛盾する．  
    &nbsp;
- `A` に文字列が代入されている場合でも，`A = A + "est"` は，`A` の値に文字列 est を連結した値を `A` の新しい値とする働きを持つ．  
  例えば，`A` の値が文字列 small ならば，`A = A + "est"` を実行した後の `A` の値は文字列 smallest になる．  
  &nbsp;
- ちなみに，「等しい」を表す記号
  (比較演算子 - 実際に等しい場合は True，等しくない場合は False の値を取る演算子) は，
  = を2つ並べた `==` である．これも間違いやすいので要注意！  
  &nbsp;
- なお，余談ではあるが，昔のプログラミング言語では，代入を `:=` で，  
  「等しい」を表す比較演算子を `=` で表すものもあった．(ALGOL，Pascal など)

以下の3つの例は，結果がどうなるか予想してからやってみよう：

````python
# 例1
A = 0
print(A)
A = A + 1
print(A)
A = A + 1
print(A)
A = A + 1
print(A)
````

````python
# 例2
B = "q"
print(B)
B = B + "w"
print(B)
B = B + "e"
print(B)
B = B + "r"
print(B)
````

````python
# 例3
A = 0
B = str(A) + ", "
A = A + 1
B = B + str(A) + ", "
A = A + 1
B = B + str(A) + ", "
A = A + 1
B = B + str(A) + ", "
print(B)
````

## プログラムの処理方法 (再掲)

1. 処理を上から順番に一つずつ実行する <u>**順次処理**</u>
2. 条件によって処理を選択する <u>**分岐処理**</u>
3. 条件が満たされている間、処理を繰り返す <u>**反復処理**</u>

- while文によるループは，この 3. <u>**反復処理**</u> に該当．
- 先に学習したif文による条件分岐は，この 2. <u>**分岐処理**</u> に該当．

## while文
<hr>

```python
while 条件式:
    反復される処理
    ...
次の処理
...
```
<hr>

注意：

- `反復される処理 ...` の部分は，(`while`，`次の処理 ...` に対して) <u>**必ず4文字下げる**</u>．
- `次の処理 ...` は，<u>**4文字戻して `while` とそろえる**</u>．
  
その動作：

- `条件式` が `True` (真) であるあいだ，`反復される処理 ...` を繰り返して実行する．
- `条件式` が `False` (偽) になると，`次の処理 ...` から処理を続行する．

## while文の例

「while文によるループ」ページの例：

````python
# 例4
text = "" # text に初期値として "" を設定する

while text != "finish": # text が "finish"ではない間、処理を繰り返す
    # 文字を入力する
    text = input("finishと入力してください:")  # text に文字を読み込む
    print(text, "と入力されました。")

print("終了しました")
````

別の例 (以下の2つの例は，結果がどうなるか予想してからやってみよう)：

````python
# 例5
A = 0
while A < 5:
    print(A)
    A = A + 1
````

````python
# 例6
B = ""
text = input("終了する際には，空文字列を入力してください:")

while text != "":
    B = B + text + ", "
    text = input("終了する際には，空文字列を入力してください:")

print(B)
````

## ループが止まらなくなってしまったら

````python
# ← ここの欄外の □ をクリックする．(例7)

i = 0           # i の初期値として 0 を与える
while True:     # 無限に処理を繰り返す (無限ループ)
    i = i + 1   # i に 1 を足した値を，新たに i の値とする

print("ここに到達する前に中断されると，ここは実行されない")

# 中断した後の変数 i の値を調べてみよう．その値が繰り返された回数を表す
````

## break文による中断

［注］「break文による中断」のページで「input()を<u>10</u>回繰り返す処理は、次のように書けます。」  
　　　 とあるのは，「input()を<u>5</u>回繰り返す処理は、次のように書けます。」の誤りです．

- ループの途中で `break` 文を実行すると，ループを抜け出せる．
    <hr>

    ```python
    while 反復条件:     # 反復条件が False の時は，ループを脱出して 次の処理 に進む
          :
        break           # ループを脱出して 次の処理 に進む
          :
    次の処理
    ```
    <hr>

    ```python
    while 反復条件:     # 反復条件が False の時は，ループを脱出して 次の処理 に進む
        処理1
        if 脱出条件:    # 脱出条件が True の時は …
            処理2       # 処理2 を実行してから …
            break       # ループを脱出して 次の処理 に進む
        処理3           # 脱出条件が False の時は 処理3 を実行し，ループを続ける
    次の処理
    ```

## continue文
- ループの途中で `continue` 文を実行すると，ループの先頭 (反復条件の判定の前) に戻る．
    <hr>

    ```python
    while 反復条件:     # 反復条件が False の時は，ループを脱出して 次の処理 に進む
          :
        continue        # ループの先頭 (反復条件の判定の前) に戻る
          :
    次の処理
    ```
    <hr>

    ```python
    while 反復条件:     # 反復条件が False の時は，ループを脱出して 次の処理 に進む
        処理1
        if 戻り条件:    # 戻り条件が True の時は …
            continue    # ループの先頭 (反復条件の判定の前) に戻る
        処理2           # 戻り条件が False の時は 処理2 を実行し，ループを続ける
    次の処理
    ```
