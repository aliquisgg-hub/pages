## 補足事項

### `len()` 関数

`len()` 関数は，文字列の長さを返します．

```python
numbers = "0123456789"
print(len(numbers))
print(len("abcdef"))
```

### 文字列の部分切り出し
※ まずは試して，法則をつかもう

```python
# 1文字
s = "abcdef"
s[0], s[1], s[5], s[-1], s[-2], s[-6]
```

```python
# 範囲(1)
s = "abcdef"
s[0:1], s[0:3], s[0:6], s[0:], s[0:-1], s[0:-3], s[:-3]
```

```python
# 範囲(2)
s = "abcdef"
s[2:5], s[1:-2], s[-5:4]
```

こう考えるとわかりやすいかも



### `capitalize()` メソッド
`capitalize()` メソッドは，先頭の英字を大文字にし，残りを小文字にします．

```python
s = "this IS a PEN!"
s.capitalize()
```

```python
s = "  this IS a PEN!  "
s.capitalize()
```

### `title()` メソッド
`title()` メソッドは，各単語の先頭の英字を大文字にし，残りを小文字にします．

```python
s = "this IS a PEN!"
s.title()
```

```python
s = "  this IS a PEN!  "
s.title()
```

### `strip()` メソッド
`strip()` メソッドは，文字列の先頭と末尾の連続する空白を全て取り除きます．

```python
s = "    this IS a PEN!    "
s.strip()
```

```python
s = "    this IS a PEN!    "
s.strip().capitalize()
```
