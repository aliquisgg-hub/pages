## リストの内包表記
- `[ ]` の中で `for … in ……` を使う

```python
# 例:
l = [1,4,2,8,5,7]
print([x for x in l])
print([2*x+1 for x in l])
print([x*x for x in l])
```

## range() 関数と for 文
決まった回数の繰り返しには，range() 関数と for 文の組み合わせが便利である

```python
for i in range(5):  # 5回繰り返す
    print(i);
```

これは，下と（ほぼ）同じ

```python
for i in [0, 1, 2, 3, 4]:
    print(i)
```

- range 関数の戻り値はrange型と呼ばれるデータ型でシーケンスの1種であり，主に for-in 文の in の後に用いられるが，リストやタプルではない
- list 関数を用いるとリスト，tuple 関数を用いると，タプルに変換できる

```python
r = range(5)   # 0, 1, 2, 3, 4 を順に「返す」シーケンス
print(r)
print(type(r))
print(tuple(r))
```

- range 関数は 3つの引数 ($start$, $stop$, $step$) を取り，$start$ から $step$ ずつの間隔で，$stop$ 未満の範囲の数値を順に返すような range 型のシーケンスを返す
- $step$ を省略したときは 1 とみなされる
- $start$ を省略したときは 0 とみなされる
- $step$ がマイナスの時は，$stop$ より大きい範囲で数値を順に返す

```python
# 分かりやすいように，タプルに変換して表示
print(tuple(range(0, 10, 1)))   # 0 から 1 ずつの間隔で，10 未満の範囲の数値を順に返すシーケンス
print(tuple(range(0, 10)))      # step が省略されている (1 とみなされる)
print(tuple(range(10)))         # start が省略されている (0 とみなされる)
print(tuple(range(1, 10)))      # step が省略されている (1 とみなされる)
print(tuple(range(0, 10, 2)))   # step が正なら stop(=10) 以上にはならない
print(tuple(range(10, 0, -2)))  # step が負なら stop(=0) 以下にはならない
print(tuple(range(0, 10, -2)))  # これは 空 になる
```

- リストの内包表記でも使える

```python
print([x*x for x in range(1,11,2)])
```

- range 型のシーケンスは，主に for-in の in の後に用いられるが，値が必要ない場合は，変数 \_ とともに用いることが多い  
  (文字 \_ は英字扱いなので，\_ 単独で変数として使える)

```python
for _ in range(5):
    print("Hello!")

print([0 for _ in range(10)])
```

- range 関数に類似した [enumerate](https://docs.python.org/ja/3.8/tutorial/datastructures.html#tut-loopidioms) 関数というのもある．
- enumerate 関数は，リストの添え字と値のペア (タプル) のシーケンスを返す．  
  ( enumerate 型．リストやタプルではない)

```python
l = ["a", "b", "c", "d", "e"]
e = enumerate(l)
print(e)
print(type(e))
print(list(e))
```

```python
l = ["a", "b", "c", "d", "e"]
for i, v in enumerate(l):
    l[i] = v*i
print(l)
```
