# タプルとコレクション の要約と補足

## タプル
- 複数のデータの組み合わせから構成されているデータを表現するオブジェクト．組．
- 複数の要素が決まった順番に並んだデータで，複数の値をカンマ `,` で区切り，通常は括弧 `()` で括って表す．
- 機能的にはリストとよく似ている（`()` と `[]` の違いはあるが…）
    - いずれも，<u>シーケンス</u>と呼ばれるデータの種類に分類される．
    - 添え字を付けて参照したり，部分的に切り出したりもできる．
    - ただし，タプルは一度生成すると，その要素の追加や削除、変更などの操作はできない．

〔三浦注〕
- お互いに強く結びついていて，全体で何か<u>ひとまとまり</u>のデータ（例えば，ばらばらにすると意味が変わってしまうようなデータ．「[ゼロからのPython入門講座](https://www.python.jp/train/tuple/index.html#%E3%82%BF%E3%83%97%E3%83%AB%E3%81%A8%E3%83%AA%E3%82%B9%E3%83%88)」では，「独立したデータ」と呼んでいる）を表すときなどに用いられる．
    - 例えば1点を表す座標値や，色を表す三原色の値の組や，データベースの「1件」のデータなど．
    - その要素を部分的に書き換えられないという性質は，全体がひとまとまりのデータであるという考えに関係している．
    - 全体で何かひとまとまりのデータを表すので，そのようなデータがたくさんある場合は，同じ構成のタプルをたくさん生成して用いることになる．
- 一方リストは，同じ種類の（ひとまとまりの）データが不定個数集まっている場合や，異なった種類のデータを単にひとくくりにしたい場合などに用いられる．

```python
l1 = [1, 2, 3]
t1 = (1, 2, 3)

print(l1[1])
print(t1[1])

l1[1] = 22
t1[1] = 22  # エラー (TypeError: 'tuple' object does not support item assignment)
```

```python
t0 = ()             # 長さ0 のタプル
t  = (1)            # 整数の1（括弧でくくられているだけ，タプルではない）
t1 = (1,)           # 長さ1 のタプル（上の様に解釈されないように，カンマを付ける）
t3 = ("a", 1, True) # 長3 のタプル

print(t0, t, t1, t3)

print(t3[0])        # 添え字で値を参照できる
print(t3[1])
print(t3[-1])
print(t3[1:-1])     # 部分（スライス）を切り出せる

t3[0] = "b"         # 代入はできない（エラーになる）
```

## コレクション，シーケンス

- コレクション
    - 複数のデータをひとくくりにするオブジェクト
    - 文字列，リスト，タプル，辞書 はいずれもコレクション
    - ひとくくりにされたデータに<u>順序があるとは限らない</u>
    - `len`関数，`==`, `!=` 演算子，`in` 演算子，for 文を用いることができる．  
      （`in` 演算子 は for 文の in とは別）
- シーケンス
    - コレクションのうち，ひとくくりにされたデータに<u>順序がある</u>もの
    - 文字列，リスト，タプル，はいずれもシーケンス
    - シーケンスはコレクションでもあるので，len関数，== 演算子，!= 演算子，in 演算子，for 文を用いることができる． 
    - 加えて，添え字（インデックス）で要素を参照したり，部分的なシーケンスを切り出したり，`<`, `<=`, `>`, `>=` 演算子で大小比較したりできる．
    

```python
str1 = "abc"
list1 = [0, 1, 2, 3]
tuple1 = (10, 20, 30)
dict1 = {"a": 1, "b": 2}

print(len(str1), len(list1), len(tuple1), len(dict1))       # ⇒ 3 4 3 2

print(str1 == "abc", str1 == "ABC")                         # ⇒ True False
print(str1 != "abc", str1 != "ABC")                         # ⇒ False True

print(list1 == [0, 1, 2, 3], list1 == [0, 1, 3])            # ⇒ True False
print(list1 != [0, 1, 2, 3], list1 != [0, 1, 3])            # ⇒ False True

print(tuple1 == (10, 20, 30), tuple1 == (10, 30))           # ⇒ True False
print(tuple1 != (10, 20, 30), tuple1 != (10, 30))           # ⇒ False True

print(dict1 == {"b": 2, "a": 1}, dict1 == {"b": 1, "a": 2}) # ⇒ True False # 順序は無関係
print(dict1 != {"b": 2, "a": 1}, dict1 != {"b": 1, "a": 2}) # ⇒ False True # 順序は無関係

print("b" in str1, "ab" in str1, "ba" in str1)              # ⇒ True True False
print(1 in list1, [1, 2] in list1, [1, 2] in [0, [1, 2], 3])# ⇒ True False True
print(20 in tuple1, 40 in tuple1)                           # ⇒ True False
print("b" in dict1, "ab" in dict1, 1 in dict1)              # ⇒ True False False
```

```python
str1 = "abc"
list1 = [0, 1, 2, 3]
tuple1 = (10, 20, 30)

print(str1[1], list1[1], tuple1[1])                                             # ⇒ b 1 20
print(str1[-1], list1[-1], tuple1[-1])                                          # ⇒ c 3 30
print(str1[1:-1], list1[1:-1], tuple1[1:-1])                                    # ⇒ b [1, 2] (20,)

print("abc" > "ab", "abc" > "abd", "abc" > "abbc")                              # ⇒ True False True
print([1, 2, 3] > [1, 2],  [1, 2, 3] > [1, 2, 4],  [1, 2, 3] > [1, 2, 2, 3])    # ⇒ True False True
print((1, 2, 3) > (1, 2),  (1, 2, 3) > (1, 2, 4),  (1, 2, 3) > (1, 2, 2, 3))    # ⇒ True False True
```

## アンパック

- 代入文の左辺に複数の変数をコンマで区切ってならべ，右辺のシーケンスを個別に代入することができる．これを<u>アンパック</u>と呼ぶ．
- 関数の戻り値をシーケンス（タプル）にすることで，複数の値を返す関数が定義できる．
  - この関数の戻り値は，構成する複数の値を，アンパックを用いて複数の変数で受け取ることができる．

```python
s0, s1, s2 = "abc"
print(s0, s1, s2)       # ⇒ a b c

l0, l1, l2 = [1, 2, 3]  # ⇒ 1 2 3
print(l0, l1, l2)

t0, t1, t2 = (1, 2, 3)  # ⇒ 1 2 3
print(t0, t1, t2)
```

```python
def shisoku(x, y):
    return (x+y, x-y, x*y, x/y)     # 括弧は省略可

wa, sa, seki, sho = shisoku(3, 3)
print(wa, sa, seki, sho)            # ⇒ 6 0 9 1.0
```
