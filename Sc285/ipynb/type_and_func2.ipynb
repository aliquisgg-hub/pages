### 変数名
- 半角のアルファベットと数字，`_` だけを使うようにしましょう

```text
variable -> OK (アルファベットのみ)
variable_123 -> OK (アルファベットと _ と数字のみ)
123_variable -> NG (数字で始まっている)
vari$able -> NG (記号は変数名として使えない)
raise -> NG (「raise」という単語はPythonで特別な意味を持つので使えない)
```

## モジュールとimport

### import文

```python
import モジュール名
```

````python
import math # mathモジュールを使用する
````

````python
import math # mathモジュールを使用する
math.sqrt(2.0) # インポート済みの mathモジュールを使って、平方根を計算する
````

````python
import calendar # calendarモジュールを使用する
calendar.prmonth(2020, 12) # 2020年１２月のカレンダーを表示する
````

### import についての補足

```python
from A import x     # モジュール A から x をインポートする
```

````python
from math import sqrt
sqrt(3)
````

````python
import math
math.sqrt(3)
````

```python
import A as B       # モジュール A を B という名前でインポートする
```

````python
import matplotlib.pyplot as plt
plt.bar(['A', 'B', 'C'], [1, 3, 2])
plt.show()
````

````python
import matplotlib.pyplot
matplotlib.pyplot.bar(['A', 'B', 'C'], [1, 3, 2])
matplotlib.pyplot.show()
````
