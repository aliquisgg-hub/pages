### [Pandas Excel、CSVファイルの読み込み、書き込み(出力)](https://ai-inter1.com/python-pandas_csv/)


```python
import pandas as pd
url = "https://raw.githubusercontent.com/AileenNielsen/TimeSeriesAnalysisWithPython/master/data/AirPassengers.csv"

df = pd.read_csv(url)
display(df)
print(df.info())
print(df.index)

df.to_csv("AP.csv")
df.to_csv("AP_noindex.csv", index=None)

with pd.ExcelWriter('AP.xlsx') as writer:
    df.to_excel(writer, sheet_name="AP")
    df.to_excel(writer, sheet_name="AP_noindex", index=None, )

df2 = pd.read_csv(url, index_col='Month')
display(df2)
print(df2.info())
print(df2.index)

df2.to_csv("AP2.csv")
with pd.ExcelWriter('AP.xlsx',engine='openpyxl', mode='a') as writer2:
    df2.to_excel(writer2, sheet_name="AP2")


```

- DataFrameへの行・列の追加と削除

```python
import pandas as pd
dict1 = {"Col1":[1,2,3], "Col2":[11,12,13], "Col3":[21,22,23]}
df4 = pd.DataFrame(dict1, index=[0,2,1])
display(df4)

print("===行の追加===")
# 1行の DateFrame を作って，列名を合わせて pd.concat() を使う
display(pd.concat([df4, pd.DataFrame([[4,14,24]], columns=df4.columns)]))
# index を振り直すなら，ignore_index=True を付ける
display(pd.concat([df4, pd.DataFrame([[4,14,24]], columns=df4.columns)], ignore_index=True))
# index の重複を避けるなら，こんな手もある
display(pd.concat([df4, pd.DataFrame([[4,14,24]], columns=df4.columns, index=[max(df4.index)+1])]))

# df を書き換えてもよければ，.loc() が使える
df4.loc[max(df4.index)+1] = [4,14,24]
display(df4)

print("===列の追加===")
# df["Col4"]=...
# df.assign(...)
# df.insert(...)
# pd.concat(..., axis=1) - pass?
```
