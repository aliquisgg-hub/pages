```python
import pandas as pd
df = pd.DataFrame([[1,2,3],[4,5,6],[7,8,9]], columns=["A","B","C"], index=[1001,1002,1003])
df
```

```python
display(df["B"])                # Series
display(df["B"][1003])          # スカラー
display(df["B"][[1001, 1003]])  # Series
display(df["B"][1:3])           # Series
```

```python
display(df[1:3])            # DataFrame
```

```python
display(df[[True, False, True]])    # DataFrame
display(df.index % 2 == 1)          # array
display(df[df.index % 2 == 1])      # DataFrame
```

```python
display(df[["B"]])                              # DataFrame
display(df[["A", "C"]])                         # DataFrame
display(df[["C", "A"]])                         # DataFrame
display(df[["C", "A"]][1:3])                    # DataFrame
display(df[["C", "A"]][[True, False, True]])    # DataFrame
```

```python
display(df.loc[1001])
display(df.iloc[1])
```
