## オブジェクトと型（クラス）

- **オブジェクト**： データ（名前が付けられるもの $\cdots$ 代入もデータに名前を付ける操作とみなせる）
- **型（クラス）**： オブジェクトの種類
  |型（クラス）|意味|説明|
  |---------|---|---|
  | int     | 整数 | 整数値を表す値．小数点なしで表現．|
  | float   | 実数 | 実数値を表す値．小数点ありで表現．|
  | str     | 文字列 | 文字の並び．引用符（" または '）で括って表現．<br>文字のシーケンス<sup>†</sup>ともみなせる．|
  | bool    | ブール値 | 真偽を表す値．(`True` か `False`)|
  | function| 関数 | 関数を表すオブジェクト． |
  | list    | リスト | データの並びを表すオブジェクト．角括弧（`[` と `]`）で括って表現．<br>個々の要素のシーケンスとみなされる．<br>*可変*（後からデータ要素の変更・追加・削除が可能）．|
  | tuple   | タプル | データの並びを表すオブジェクト．丸括弧（`(` と `)`）で括って表現．<br>個々の要素のシーケンスとみなされる．<br>*不可変*（後からデータ要素の変更・追加・削除ができない）．||
  | dict    | 辞書 | キー（見出し）と値のペアの集まり．波括弧（`{` と `}`）とコロン（`:`）を用いて表現．<br>キーを用いて値をアクセスするコレクション<sup>‡</sup> とみなされる．<br>*可変*（後からデータ要素の変更・追加・削除が可能）．|
  | module  | モジュール | インポートされたモジュールを表すオブジェクト．<br>さまざまな関数（メソッド）や値（プロパティ）などを内部に含む．|

  ※ 上記以外にもたくさんある（自分で定義することもできる）  
  ※ type関数で調べることができる．

  <sup>†</sup> シーケンスはデータの並びをひとまとめにしたデータのこと（コレクションの1種ともみなせる）  
  <sup>‡</sup> コレクションはデータの集まりをひとまとめにしたデータのこと（シーケンスも包含する）



```python
import math                 # math のインポート
def printType(x):           # x の型（クラス）を表示する関数
    print(f'{repr(x)} : {type(x)}')
# repr(x) の値は，オブジェクト x の文字列表現（x が文字列の場合，引用符付きになる）
# type(x) の値は，オブジェクト x のクラスを表す type オブジェクト

l0 = [1,2]                  # 整数のリスト (list)
t0 = (1,2)                  # 整数のタプル (tuple)
d0 = {'one':1,'two':2.0}    # 文字列 (str) をキーとする辞書 (dict)

printType(1)            # : <class 'int'>
printType(1.0)          # : <class 'float'>
printType(math.pi)      # math.pi==3.141592653589793 : <class 'float'>
printType('1')          # : <class 'str'>
printType(True)         # : <class 'bool'>
printType(printType)    # printType==<function printType at ...> : <class 'function'>
printType(math.sqrt)    # math.sqrt==<built-in function sqrt> : <class 'builtin_function_or_method'>
printType(l0)           # l0==[1,2] : <class 'list'>
printType(l0[0])        # l0[0]==1 : <class 'int'>
printType(t0)           # t0==(1,2) : <class 'tuple'>
printType(t0[1])        # t0[1]==2 : <class 'int'>
printType(d0)           # d0=={'one':1<class ','two':2<class '.0} : <class 'dict'>
printType(d0['one'])    # d0['one']==1 : <class 'int'>
printType(d0['two'])    # d0['two']==2.0 : <class 'float'>
printType(math)         # math==<module 'math' (built-in)> : <class 'module'>

# 1, 1.0, '1', True などは，値を表す定数（リテラル）である
# math, printType, l0, t0, d0 は値を持つ変数（のようなもの）と考えるとよい
# math.pi, math.sqrt は math モジュールのメンバー（変数のようなもの）である

printType(type(1))      # type(1)==<class 'int'> : <class 'type'>
```

### 型（クラス）の変換
- 型（クラス）の名前のうち，`int`, `float`, `str`, `bool`, `list`, `tuple`, `dict` 等は，型（クラス）を変換する関数のように用いることができる  
  （これら `int`, `float`, `str`, `bool`, `list`, `tuple`, `dict` 等の実体は，各型（クラス）に対応する type オブジェクトである）
- 変換の個々のケースについて，紛らわしい場合はまず試してみること

```python
printType(int)          # int==<class 'int'> : <class 'type'>
print()                 # 空行をあける
print(int(2.0))         # float → int   (2)
print(int('-12'))       # str   → int   (-12)
print(int(True))        # bool  → int   (1)
print()                 # 空行をあける
print(float(12))        # int   → float (12.0)
print(float('-12'))     # str   → float (-12.0)
print(float(True))      # bool  → float (1.0)
print()                 # 空行をあける
print(repr(str(12)))    # int   → str   ('12')
print(repr(str(2.0)))   # float → str   ('2.0')
print(repr(str(True)))  # bool  → str   ('True')
print()                 # 空行をあける
print(bool(0))          # int   → bool  (False)
print(bool(0.0))        # float → bool  (False)
print(bool(''))         # str   → bool  (False)
```

```python
print(repr(str([1,2])))             # list  → str       '[1, 2]'
print(repr(str((1,2))))             # tuple → str       '(1, 2)'
print(repr(str({'A':1, 'B':2})))    # dict  → str       "{'A': 1, 'B': 2}"
print()                             # 空行をあける
print(list('ABC'))                  # str   → list      ['A', 'B', 'C']
print(list((1,2)))                  # tuple → list      [1,2]
print(list({'A':1, 'B':2}))         # dict  → list      ['A', 'B']
print()                             # 空行をあける
print(tuple('ABC'))                 # str   → tuple     ('A', 'B', 'C')
print(tuple([1,2]))                 # list  → tuple     (1,2)
print(tuple({'A':1, 'B':2}))        # dict  → tuple     ('A', 'B')
print()                             # 空行をあける
print(dict([['A',1], ('B',2)]))     # list  → dict      {'A': 1, 'B': 2}
print(dict((('A',1), ['B',2])))     # tuple → dict      {'A': 1, 'B': 2}
# ↑ 要素の2個組は list でも tuple でもよい
print()                             # 空行をあける
print(bool([]), bool([0]))          # list  → bool      False True
print(bool(()), bool((0,)))         # tuple → bool      False True
print(bool({}), bool({0:0}))        # dict  → bool      False True
```

## オブジェクトのリテラル表現（定数の表現）

- **整数 (int)**: 小数点なしの10進数（先頭が0でない数字の並び），前に符号をつけることができる
  - 2進数，8進数，16進数の表現もあるが省略
- **実数 (float)**: 小数点を含む10進数，前に符号をつけることができる
  - `3.14e3` のような，指数表現も可能 (この例は $3.14×10^3$ を表す)
- **文字列 (str)**: 引用符(' または ")で括った文字の並び
  - 最初の引用符の直前に `f` を付けると，**フォーマット済み文字列リテラル**として扱われる  
  - フォーマット済み文字列リテラルの中には，波括弧で括って式を含めることができる
- **ブール値 (bool)**: `True`, `False` で表す

```python
print(1, -2, +3)                    # 整数リテラルの例
print(1.0, 1., -1.22, -0.33, -.33)  # 実数リテラルの例
print(2e3, 2e+3, 2.1e+3, -2.1e-3)   # 実数リテラルの例（指数表現）
print('ab', "cd", 'e"f', "g'h")     # 文字列リテラルの例
print("e\"f", 'g\'h')               # 文字列リテラルの例
x=2.5; y=3; z="cm"                  # 複数の命令文を ; で区切って1行にまとめることができる
print(f'{x}{z} * {y} = {x*y}{z}')   # フォーマット済み文字列リテラルの例
print(True, False)                  # ブール値リテラルの例
```

## 構造を持つオブジェクトの直接的表現

- **リスト (list)**: 任意個数の要素 (式) をコンマ (`,`) で区切って，角括弧 (`[` `]`) で括って表す
- **タプル (tuple)**: 任意個数の要素 (式) をコンマ (`,`) で区切って，丸括弧 (`(` `)`) で括って表す
  - 要素数が1個の時は，その要素の後ろにコンマ (`,`) を付けて表す
- **辞書 (dict)**: <u>キー</u> (式) と<u>値</u> (式) をコロン (`:`) 
  で区切った組を，コンマ (`,`) で区切って並べて，波括弧 (`{` `}`) で括って表す

&emsp;※ 式については次項

```python
print([], [1], [1, 2.0, 'three', False, [1,2], (1,2)])  # リストの例（[] は空リスト）
print((), (1,), (1, 2.0, 'three', False, [1,2], (1,2))) # タプルの例（() は空タプル）
print((1))                                              # (1) は 1 と同じでタプルではない

print({'A':1, 'B':2.0, 3:['C',3]})      # 辞書の例
```

## 式 ― プログラムの中で値を持つ表現

- プログラムの中で，値を持つ（値を計算する）表現を式という
- 以下のようなものが式である．
  - オブジェクトのリテラル表現（上述）― `1`, `2.0`, `'three'`, `False` など
  - 構造を持つオブジェクト（リスト，タプル，辞書）の直接的表現（上述）― `[1,2]`, `(3,4)`, `{'A':5, 'B':6}` など
  - 変数（代入されている値が式としての値となる）
  - 関数名や（インポートされたモジュールの）モジュール名  
    ― `max`, `len`,（`import math` を実行した後の）`math`など  
    &emsp;（その関数やモジュールそのもの（オブジェクト）が式としての値となる）
  - モジュールのメンバー：モジュールを表す式の後に，ピリオド（`.`）をはさんでメンバー名を続けたもの  
    ― （`import math` を実行した後の）`math.pi`（実数値）や `math.sqrt`（関数）など
  - シーケンス（文字列，リスト，タプル）の要素や部分を表す表現（詳細は後述）
    - シーケンスを表す式の後に，角括弧（`[` と `]`）で括ったインデックス（添え字）を付けたもの  
      ―（`x` の値がシーケンスのとき）`x[0]` など
    - シーケンスを表す式の後に，初めと終わりのインデックスをコロン（`:`）で対にして，角括弧（`[` と `]`）で括って付けたもの  
      ―（`x` の値がシーケンスのとき）`x[1:3]` など
  - 辞書を表す式の後に，角括弧（`[` と `]`）で括ったキーを付けたもの  
    ―（`x` の値が辞書のとき）`x[’A']` など（詳細は後述）
  - 関数呼び出し：関数を表す式の後に，丸括弧（`` と ``）で括った引数を付けたもの  
    ― `max(1,4,2,3)`, `len([1,3,5])`, `math.sqrt(2)` など
  - メソッド呼び出し：オブジェクトを表す式の後に，ピリオド（`.`）をはさんでメソッド名と丸括弧（`` と ``）で括った引数を続けたもの  
    ― `'DataFrame'.lower()`, `'DataFrame'.count('a')` など
  - 式を演算子で組み合わせたもの ― `x+1`, `b**2-4*a*c`, `'OK' if x==0 else 'NG'`（`if` と `else` がセットになった3項演算子） など
- 値を持たない式もある（実際は，None という値を持っている）
  - 例えば `print('abc')`, （`x` の値がリストのとき）`x.sort()` など

```python
print(1, 2.0, 'three', False)           # オブジェクトのリテラル表現
print([1,2], (3,4), {'A':5, 'B':6})     # 構造を持つオブジェクト（リスト，タプル，辞書）の直接的表現
x = 3.14                                # 実数
print(x)                                # 変数
import math
print(max, len, math)                   # 関数名や（インポートされたモジュールの）モジュール名
print(math.pi, math.sqrt)               # mathモジュールのメンバー
x = ('零','壱','弐','参')               # タプル
print(x[0], x[1:3])                     # シーケンス（タプル）の要素や部分を表す表現
x = {'A':1, 'B':2.0, 3:['C',3]}         # 辞書
print(x['A'], x[3])                     # 辞書を表す式の後に，角括弧（`[` と `]`）で括ったキーを付けたもの
print(max(1,4,2,3), len([1,3,5]), math.sqrt(2))     # 関数呼び出し
print('DataFrame'.lower(), 'DataFrame'.count('a'))  # メソッド呼び出し
x=1; a=1; b=4; c=2
print(x+1, b**2-4*a*c, 'OK' if x==0 else 'NG')      # 式を演算子で組み合わせたもの
```

### 演算子のまとめ

よくわからない演算子もあると思うが，必要になったときに調べればよい．今は気にしないでおこう．
| 演算子 |説明|
|:------:|:---|
|*x* `+` *y* |1.（数値）*x* と *y* の和<br>2.（シーケンス）*x* と *y* の連結|
|*x* `-` *y* |（数値）*x* と *y* の和|
|*x* `*` *y* |1.（数値）*x* と *y* の積<br>2.（シーケンス×整数）*x* の複製を *y* 個連結|
|*x* `/` *y* |（数値）*x* と *y* の商（結果は float）|
|*x* `//` *y*|（数値）*x* と *y* の商の小数点以下切り捨て|
|*x* `%` *y* |（数値）*x* を *y* で割った剰余|
|*x* `**` *y*|（数値）*x* の *y* 乗|
|`-`*x*      |（数値）*x* の符号を反転|
|||
|*x* `\|` *y*|（数値）*x* と *y* のビット単位の論理和|
|*x* `^` *y* |（数値）*x* と *y* のビット単位の排他的論理和|
|*x* `&` *y* |（数値）*x* と *y* のビット単位の論理積|
|`~`*x*      |（数値）*x* の全ビットを反転|
|*x* `<<` *y*|（数値）*x* を *y* ビット左シフト|
|*x* `>>` *y*|（数値）*x* を *y* ビット右シフト|
|||
|*x* `==` *y*|*x* と *y* が等しい|
|*x* `!=` *y*|*x* と *y* が等しくない|
|*x* `>` *y* |*x* が *y* より大きい（シーケンスの場合辞書順）|
|*x* `<` *y* |*x* が *y* より小さい（シーケンスの場合辞書順）|
|*x* `>=` *y*|*x* が *y* 以上（シーケンスの場合辞書順）|
|*x* `<=` *y*|*x* が *y* 以下（シーケンスの場合辞書順）|
|*x* `is` *y*|（リテラルでないオブジェクト）*x* と *y* が同一オブジェクトである|
|*x* `is not` *y*|（リテラルでないオブジェクト）*x* と *y* が同一オブジェクトでない|
|*x* `in` *y*|1.（*y* がシーケンス）*x* は *y* の要素である<br>2.（*y* が辞書）*x* は *y* のキーである|
|*x* `not in` *y*|1.（*y* がシーケンス）*x* は *y* の要素でない<br>2.（*y* が辞書）*x* は *y* のキーでない|
|||
|*x* `or` *y* |*x* が偽でない（偽とみなせない<sup>†</sup>）なら *x*，そうでなければ *y*|
|*x* `and` *y*|*x* が偽（とみなせる<sup>†</sup>）なら *x*，そうでなければ *y*|
|`not` *x*    |*x* が偽（とみなせる<sup>†</sup>）なら `True`，そうでなければ `False`|
|*x* `if` *y* `else` *z*|*y* が偽でない（偽とみなせない<sup>†</sup>）なら *x* そうでないなら *z*|

&emsp;<sup>†</sup>`False` 以外に，`0`, `0.0`, `''`, `[]`, `()`, `{}`, `None` 等も偽とみなせる
