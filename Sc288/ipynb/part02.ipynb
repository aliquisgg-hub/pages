## 関数の定義と呼び出し

- 関数は以下のように定義する  
  ( `処理1`, `処理2`, ... という内容を，`関数名` という名前で定義する)．

  ```python
  def 関数名(引数名1, 引数名2, ...):
      処理1
      処理2
      ...
  ```

- 定義内容 (本体とも呼ばれる) ( `処理1`，`処理2`, ...) は<span style="color:red;">**4字下げる**</span>．
- `引数名1`, `引数名2`, ... は，**仮引数**と呼ばれ，関数を呼び出した際に値が代入されるローカル変数 (後述) のように扱われる．
- 関数を呼び出す際には，関数名の後の括弧の中に，仮引数に代入する値のための式を並べる．これを<u>実引数</u>という．
- 関数は，単独で文として呼び出されたり，式の一部として (値を持った部分式として) 呼び出されたりする．
- 関数の**戻り値**（関数を式の中で呼び出した時の式としての値）は，**retrun 文**を用いて与える．

  ```python
  return 値
  ```
  - 関数の本体で retrun 文が実行されると，関数の実行を直ちに中断し，与えられた値を戻り値として，呼び出されたところに戻る．(break と少し似ているかもしれない)
- 値が必要ない関数 (文として呼び出され，式の一部としては呼び出されない関数) の場合，return 文は必要ないが，式の一部として呼び出される関数は，必ず return 文で戻らなければならない．
  - 実は …… return 文のない関数の戻り値や，`値` のない `return` だけの
    retrun 文で戻った場合の戻り値は，`None` という値（「値なし」という意味の値，NoneType
    クラスのオブジェクト）になる．

````python
def hello1(who):
    msg = "Hello " + who    # "Hello " と who の内容を連結して msg に代入
    print(msg)              # msg を表示
                            # 戻り値は None（値なし）

def hello2(who):
    msg = "Hello " + who    # "Hello " と who の内容を連結して msg に代入
    print(msg)              # msg を表示
    return                  # 戻り値は None（値なし）

def hello3(who):
    msg = "Hello " + who    # "Hello " と who の内容を連結して msg に代入
    return msg              # 戻り値は msg の内容（文字列）

x1 = hello1("everyone")     # hello1 の実行中に Hello everyone を表示，戻り値は None
print(f'{repr(x1)} : {type(x1)}')
x2 = hello2("everyone")     # hello2 の実行中に Hello everyone を表示，戻り値は None
print(f'{repr(x2)} : {type(x2)}')
x3 = hello3("everyone")     # 戻り値は Hello everyone という文字列 (str)
print(f'{repr(x3)} : {type(x3)}')
````

## ローカル変数とグローバル変数

- **ローカルスコープ** - 関数定義の中
  - ここで代入された変数 (**ローカル変数**)，<u>定義された関数</u>，<u>import されたモジュール</u>は，このローカルスコープ内でのみ使用可能．
  - ローカルスコープとなっている関数が呼び出されるごとに，そのローカルスコープ内でのみ有効な変数 (等) が作られ，関数の実行が終了すると消去される．

````python
# 円の半径 (radius) から，円周と面積を表示する関数
#   radius    : 仮引数（ローカル変数）
#   math      : ローカルにインポートされたモジュール
#   enshu     : ローカル変数
#   circleArea: ローカルに定義された関数
def test(radius):
    import math
    enshu = 2 * math.pi * radius    # ローカル変数
    def circleArea(r):
        return math.pi*r*r
    print(enshu)
    print(circleArea(radius))

test(10)                        # 正しく機能する

try: print(enshu)               # enshu が有効ではないのでエラーになる
except Exception as e: print(e)

try: print(math.pi)             # math が有効ではないのでエラーになる
except Exception as e: print(e)

try: print(circleArea(5))       # circleArea が有効ではないのでエラーになる
except Exception as e: print(e)

# 注）
#   try:〈文〉
#   except Exception as e: print(e)
# 
#   は，〈文〉を実行してエラーが発生しても，
#   エラーメッセージを表示するだけで実行を続行する
#   (これは覚えなくてもよい)
````

- **モジュールスコープ** - ファイル内で関数定義以外の部分
  - ここで代入された変数 (**グローバル変数**)，<u>定義された関数</u>，<u>import されたモジュール</u>は，このファイル内 (モジュール内) のどこでも (関数定義の中でも) 使用できる．
  - ただし，グローバル変数にローカルスコープで代入することはできない．(代入しようとすると，同じ名前のローカル変数が別に作られる － 下の例の `enshu` を見よ)

````python
import math             # グローバルにインポートされる

def circleArea(r):      # グローバルに定義される
    return math.pi*r*r

radius = 10             # グローバル変数
enshu = 0               # グローバル変数

# 円の半径 (radius: グローバル変数) から，円周と面積を表示する関数
#   enshu   : ローカル変数（グローバルな enshu とは別に作られる）
def test():
    enshu = 2 * math.pi * radius    # 新たに enshu というローカル変数が作られる
    print(enshu)
    print(circleArea(radius))

test()                          # 正しく機能する

try: print(enshu)               # グローバル変数の方の enshu の値 (0) が表示される
except Exception as e: print(e)

try: print(math.pi)             # 正しく機能する（円周率が表示される）
except Exception as e: print(e)

try: print(circleArea(5))       # 正しく機能する（半径5の円の面積が表示される）
except Exception as e: print(e)
````

## 関数の呼び出し方

- 引数指定の2つの方法
  - **位置引数** - 実引数の順序を仮引数の順序に合わせる．
  - **キーワード引数** - 仮引数の名前を指定して実引数を与える．

````python
def func1(arg1, arg2, arg3):
    print("arg1は:", arg1, ", arg2は:", arg2, ", arg3は:", arg3)

func1(1, 2, 3)                  # 位置引数
func1(arg1=1, arg2=2, arg3=3)   # キーワード引数
func1(arg2=2, arg1=1, arg3=3)   # キーワード引数の順序は任意
func1(1, arg3=3, arg2=2)        # 混在
````

- デフォルト引数 (引数のデフォルト値) - 呼び出し時に実引数を省略した場合に仮引数に与える値
  - 関数定義の際の仮引数に `引数名=デフォルト値` という形で設定する．

````python
def func2(arg1, arg2=22, arg3=33):
    try: print("arg1は:", arg1, ", arg2は:", arg2, ", arg3は:", arg3)
    except Exception as e: print(e)

func2(1, 2, 3)          # 省略なし
func2(1, 2)             # 第3引数 (arg3) を省略
func2(1)                # 第2引数 (arg2) と第3引数 (arg3) を省略

#func2()                # エラーになる（デフォルト値を設定していない引数は省略できない）
#func2(1, , 3)          # エラーになる（この形で，第2引数 (arg2) を省略することはできない）

func2(1, arg3=3)        # 第2引数 (arg2) だけを省略
func2(arg3=3, arg1=1)   # 第2引数 (arg2) だけを省略
````

## 構造化プログラミングと制御構文

### 流れ図

プログラムが実行するべき仕事の手順を**流れ図**（**フローチャート**）を用いて表すと，プログラムの構造を把握しやすくなる．

この流れ図（フローチャート）とは，<u>処理の内容を書いた箱</u>（**図形要素**）を， <u>実行順に線（または矢印）でつないで表したもの</u>である． 一般的には，上から下・左から右へ進行するように書くが， 必要に応じて任意の方向に矢印を伸ばしても良い．

処理の内容を書いた箱（図形要素）は，その処理内容に応じて下図のような種類があるが，自分用の覚書として用いる場合は厳密に使い分けなくてもよい．（「開始・終了」「通常の処理」「分岐」の3つがあれば，大抵は何とかなる）

### 構造化プログラミング

プログラムの実行順序の構造は，すべて3つの基本的な処理の構造（**順次処理**・**分岐処理**・**反復処理**）の組み合わせで表現ができる．（構造化定理）

プログラムで実現しようとしている（複雑な）機能を，上記の3つの基本的な構造を当てはめながらより単純な機能の組み合わせとして，段階的に設計していくプログラミング手法を**構造化プログラミング**という．

流れ図で考えると，はじめはプログラム全体を一つの（四角の）箱と考え，その箱を段階的に，順次・分岐・反復のいずれかの構造で置き換えていくことで，流れ図を作り上げていくことに相当する．（参考： 「構造化プログラミングのパターン」（次項）．また，下の「構造化プログラミングの例（計算練習1）」も見よ）

十分に細分化され，分解された各々の処理が1命令で実現できるくらい単純になれば， その単純な処理を行う命令を，分解の逆をたどって組み立てていくことで， プログラムのコード全体を構成することができる．（この作業（プログラムのコードを記述すること）を*コーディング*という）

このような手法を取ることにより，プログラムの構造が整理されて把握しやすくなり，また，読みやすいプログラムを作成することができる．

<span style="font-family:sans-serif;">(*)</span>〔反復処理〕の2つ目のパターンの補足説明
- 決まった回数（例では $n$回）の反復に用いられる．
  - 変数 `i` の値を $0\ ～\ n-1$ と変えながら反復する，と考えることもできる．
- 変数は任意の名前でよい（`i` でなくてもよい）．
- 変数の初期値（例では `0`），反復条件（例では `i<n`），反復ごとの変数の更新  
  （例では `i←i+1`）は，さまざまなバリエーションがあり得る．
</div>

- 各コードパターンの制御構文については後述する
- `fo-in` を用いた反復処理のパターンについては後述する．

#### 段階的詳細化の例（計算練習1）
##### プログラムの概要

- 1～10 の自然数の足し算の問題を出題して， 解答を入力させ，正誤をチェックするプログラムを作成する．
- 問題は10問出題するものとする．
- 問題は，random.randint 関数 を用いてランダムに生成する．
- 問題は input 関数を用いて出題し， 答えを入力させる．
- 正誤の判定結果は print 関数を用いる．

(a)「10回の計算練習」を 1つの処理と考えて，箱1つの流れ図を書く．  
(b)「10回の計算練習」を，「1回の計算練習」を10回反復する処理と考える．  
&emsp;（反復処理の2つ目，n回反復のパターン．下図の緑の枠）  
(c)「1回の計算練習」を，「問題作成」「出題と回答入力」「正誤判定」の順次処理と考える．  
&emsp;（順次処理のパターン．下図の赤の枠）  
(d)「正誤判定」を正解か否かの条件によって分岐する処理と考える．  
&emsp;（分岐処理の2つ目のパターン．下図の青の枠）

### 制御構文

#### if 文（分岐処理）


- else 節がない（elif 節もない）場合
    <div style="border:solid gray 1px;">

    ```Python
    if 条件式:
        処理1
        処理2
        ...
    ```
    </div>

- else 節があり，elif 節がない場合
    <div style="border:solid gray 1px;">

    ```python
    if 条件式:
        処理1
        処理2
        ...
    else:
        処理3
        処理4
        ...
    ```
    </div>

- elif 節も else 節もある場合
    <div style="border:solid gray 1px;">
    
    ```python
    if 条件式1:
        処理1
        ...
    elif 条件式2:
        処理2
        ...
    elif 条件式3:
        処理3
        ...
    ...
    else:
        処理n
        ...
    ```
    </div>
- elif 節があり，else 節がない場合
    <div style="border:solid gray 1px;">

    ```python
    if 条件式1:
        処理1
        ...
    elif 条件式2:
        処理2
        ...
    elif 条件式3:
        処理3
        ...
    ```
    </div>

- いずれの場合も <u>`if`, `elif`, `else` の先頭の位置（開始位置）はそろえて</u>，
  <u>`処理` はインデント（字下げ）する</u>こと（<span style="color:red;">**通常は4文字分字下げする**</span>）
- `if` あるいは `elif` 直後の <u>`条件式` を順に計算</u>し，
  **最初に `True` になった条件の直後の `処理`（インデントされている範囲）のみを実行する**
  - 全ての `条件式` が `False` だった場合：  
    `else` 節があれば，その `else` 節の `処理` のみを実行する  
    `else` 節がなければ，何もしない
- `if` の次の文（続きの文）は，<u>`if`, `elif`, `else` の先頭の位置（開始位置）と同じ位置から始める</u>

````python
# 例
import random
rnd = random.randint(-2, 2)     # -2 ≦ rnd ≦ 2
print(rnd)
if rnd < 0:
    print("負の数")
````

````python
# 例
import random
rnd = random.randint(-2, 2)     # -2 ≦ rnd ≦ 2
print(rnd)
if rnd < 0:
    print("負の数")
else:
    print("非負の数")
````

````python
# 例
import random
rnd = random.randint(-2, 2)     # -2 ≦ rnd ≦ 2
print(rnd)
if rnd < 0:
    print("負の数")
elif rnd > 0:
    print("正の数")
else:
    print("ゼロ")
````

````python
# 例
import random
rnd = random.randint(-2, 2)     # -2 ≦ rnd ≦ 2
print(rnd)
if rnd < 0:
    print("負の数")
elif rnd > 0:
    print("正の数")
````

#### while 文（反復処理）

<div style="border: solid gray 1px;">

```python
while 条件式:
    反復される処理
    ...
次の処理
```
</div>

注意：

- `反復される処理 ...` の部分は，(`while`，`次の処理 ...` に対して) <u>**必ず4文字下げる**</u>．
- `次の処理 ...` は，<u>**4文字戻して `while` とそろえる**</u>．
  
その動作：

- `条件式` が `True` (真) であるあいだ，`反復される処理 ...` を繰り返して実行する．
- `条件式` が `False` (偽) になると，`次の処理 ...` から処理を続行する．


````python
# 例
n = 16
while n > 0:
    print(n)
    n = n // 2
````

````python
# 例
s = 'python'
w = len(s)      # s が 'python' なら，w はその長さ 6
i = 0
while i < w:    # w 回の反復 (i ← 0 ～ w-1)
    print(s[i])
    i = i + 1
````

````python
# 例
s = 'python'
w = len(s)      # s が 'python' なら，w はその長さ 6
j = w
while j > 0:    # w 回の反復 (j ← w ～ 1)
    print(s[j-1])
    j = j - 1
````

#### break 文による中断


- 反復処理の途中で `break` 文を実行すると，ループを抜け出せる．

    <div style="border: solid gray 1px;">

    ```python
    while 反復条件:     # 反復条件が False の時は，ループを脱出して 次の処理 に進む
          :
        break           # ループを脱出して 次の処理 に進む
          :
    次の処理
    ```
    </div><br>
    <div style="border: solid gray 1px;">

    ```python
    while 反復条件:     # 反復条件が False の時は，ループを脱出して 次の処理 に進む
        処理1
        if 脱出条件:    # 脱出条件が True の時は …
            処理2       # 処理2 を実行してから …
            break       # ループを脱出して 次の処理 に進む
        処理3           # 脱出条件が False の時は 処理3 を実行し，ループを続ける
    次の処理
    ```
    </div>

    `反復条件` が `True`（無条件の反復）で，`処理2` がない場合は……
    <div style="border: solid gray 1px;">

    ```python
    while True:         # 常に反復
        処理1
        if 脱出条件:    # 脱出条件が True の時は …
            break       # ループを脱出して 次の処理 に進む
        処理3           # 脱出条件が False の時は 処理3 を実行し，ループを続ける
    次の処理
    ```
    </div>

````python
# 例
sum = 0
while True:         # 常に反復
    n = int(input('input non-zero integer (or 0 to stop)')) # 整数を n に入力
    if n==0: break  # 入力が 0 ならループを脱出する
    sum = sum + n   # sum に n を足してループを続ける
print(sum)          # sum を表示
````

※ for-in 文による反復処理（次回解説）でも break 文は使用可能．

#### continue 文による反復の継続

- ループの途中で `continue` 文を実行すると，ループの先頭 (反復条件の判定の前) に戻る．
    <div style="border: solid gray 1px;">

    ```python
    while 反復条件:     # 反復条件が False の時は，ループを脱出して 次の処理 に進む
          :
        continue        # ループの先頭 (反復条件の判定の前) に戻る
          :             # continue した場合は，ここは実行されない
    次の処理
    ```
    </div><br>
    <div style="border: solid gray 1px;">

    ```python
    while 反復条件:     # 反復条件が False の時は，ループを脱出して 次の処理 に進む
        処理1
        if 戻り条件:    # 戻り条件が True の時は …
            continue    # ループの先頭 (反復条件の判定の前) に戻る (処理2は実行されない)
        処理2           # 戻り条件が False の時は 処理2 を実行し，ループを続ける
    次の処理
    ```
    </div>

````python
# 例
import math
while True:
    n = float(input("input positive number (or 0 to stop)"))
    if n == 0.0: break
    if n < 0.0: continue
    print(math.sqrt(n), flush=True)
print('stop')
````

※ for-in 文による反復処理（次回解説）でも continue 文は使用可能．

#### **[補足]** else 節のある while 文

- 通常の `while` 文の末尾に `else` 節を付けることができる
    <div style="border: solid gray 1px;">

    ```python
    while 反復条件:     # 反復条件が False の時は，ループを脱出して else節 に進む
            :
        break           # ループを脱出して 次の処理 に進む
            :
    else:
        反復が，(break で中断されずに) 反復条件が False で終了したときの処理
    次の処理
    ```
    </div>

````python
# 例
print('この言語の名前のスペルを1文字ずつ入力してください', flush=True)
s = 'python'
w = len(s)
i = 0
while i < w:
    c = input()
    print(c, flush=True)
    if (c != s[i]): break   # 途中で間違えるとそこで終了
    i = i + 1
else:                       # 最後まで反復したら，…
    print('正解！')         # 正解！と表示して終了
print('終了')
````

- `else` 節を用いると，多重ループの脱出（break）ができる
    <div style="border: solid gray 1px;">

    ```python
    while 条件1:        # (外の反復) 条件1が False の時は，「次の処理」に進む
        while 条件2:    # (内の反復) 条件2が False の時は，else 節に進む
                :
            break       # 内の反復を脱出して，下の break に進む
                :
        else:
            continue    # 内の反復を通常終了したとき実行され，外の反復を継続する
        break           # 内の反復を break した時実行され，外の反復も脱出する（次の処理へ進む）
    次の処理
    ```
    </div>

※ for-in 文による反復処理（次回解説）でも else 節は使用可能．

Ⅰ. `x = input()` を用いて文字列を入力し，`print(x, flush=True)` でおうむ返しにするプログラムを書いてみよう．ただし：
-   空文字列 `''` が入力されるまで，おうむ返しを繰り返すこと ．
-   `break` を使わずに書くこと．

※ `flush=True` は，引数を**ただちに**表示するためのおまじない．

ヒント：
-   最初に `x = input()` で文字列を入力し，`x` が空文字列でなければ `print(x, flush=True)` に続けて `x = input()` を実行するような反復とする．
-   何らかの条件に基づく反復は，while 文を用いる．

Ⅱ. `x = input()` を用いて文字列を入力し，`print(x, flush=True)` でおうむ返しにするプログラムを書いてみよう．ただし：
-   空文字列 `''` が入力されるまで，おうむ返しを繰り返すこと ．
-   `x = input()` を一つしか書かないこと．(`break` を使ってもよい)

※ `flush=True` は，引数を**ただちに**表示するためのおまじない．

ヒント：
-   `x = input()` に続けて `print(x, flush=True)` を実行するような反復を作成し，`x = input()` の直後に，`x`
    が空文字列 `''` の時に `break` するようなプログラムとする
-   無条件の反復は，条件が常に True になるような while 文を用いる
