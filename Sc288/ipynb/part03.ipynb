## **[補足]** 辞書の操作

### 要素の参照

-   辞書の要素は，**キー**を用いて参照する．  
    (数値，文字列，真理値，タプルはキーとして使える．リストや辞書はキーにできない)

````python
d0 = {'A':1, 2:2.0, True:999, (1, 1):[1.414, 3.14/4]}
print(d0['A'], d0[2], d0[True], d0[(1, 1)])
````

-   キーのリストは `list(辞書.keys())` で得られる
-   値のリストは `list(辞書.values())` で得られる
-   キーと値の2個組 (タプル) のリストは `list(辞書.items())` で得られる

````python
print(list(d0.keys()))
print(list(d0.values()))
print(list(d0.items()))
````

### 要素の置き換え(代入) 

-   要素の置き換えは，要素の参照に代入する形で行える

````python
print(d0)
d0['A'] = 100
d0[2] = 200.0
d0[True] = 999999
d0[(1,1)] = [1.41421356, 3.14159265/4]
print(d0)
````

### 要素の追加・削除

-   要素の追加は，まだ存在しないキーに値を代入することで行える
-   要素の削除は，その要素の参照を `del(…)` 関数の引数に与えることで行える

````python
d0 = {'A':1, 2:2.0, True:999, (1, 1):[1.414, 3.14/4]}
print(d0)
d0['B'] = 300       # 'B' に対応する要素を追加
print(d0)
del(d0['A'])        # 'A' に対応する要素を削除
print(d0)
````

## シーケンスの操作

### 要素の参照  

- シーケンスを表す式（シーケンスが代入されている変数など）の後に，インデックス（添字）を角括弧 `[]` で括って添える．
  - インデックス（添字）は整数値で，要素の順番（先頭の要素を 0 とする）を表す
  - 負数のインデックスも使用できる（末尾の要素を -1 とする）
  - コロンを使って，部分シーケンスの切り出しもできる

<div style="margin-left: 4em;">



</div>

````python
s1 = '142857'
l1 = [1,4,2,8,5,7]
t1 = (1,4,2,8,5,7)

print(len(s1),len(l1),len(t1))          # シーケンスの長さ
print([s1[1],l1[1],t1[1]])              # 左から2つ目の要素
print([s1[-1],l1[-1],t1[-1]])           # 右端の要素
print([s1[1:5],l1[1:5],t1[1:5]])        # 範囲の指定
print([s1[1:],l1[1:],t1[1:]])           # 1: は 1の位置から末尾まで
print([s1[:5],l1[:5],t1[:5]])           # :5 は 先頭から5の位置まで（0:5 と同じ）
print([s1[1:-1],l1[1:-1],t1[1:-1]])     # 1:-1 は（この例では） 1:5 と同じ
print([s1[-5:-1],l1[-5:-1],t1[-5:-1]])  # これも（この例では） 1:5 と同じ
````

### リストの要素(または部分リスト)の置き換え(代入) 

- list（リスト）ではできるが，str（文字列），tuple（タプル）ではできない
- インデックスによる要素(または部分リスト)の参照に新たな値を代入する
- 単一の要素の置き換えだけではなく，部分リスト（範囲）の他のリスト（空リストを含む任意の長さのリスト）への置き換えも可能

````python
# 1行に複数の命令(文)を並べるときには ; で区切る
l2 = [1,3,5,7,9]
l2[1]=33; print(l2)             # 1番の要素（左から2番目の要素）を置き換える

l2[2:-1]=[50,70]; print(l2)     # 2の位置 ～ −1の位置 の範囲を置き換える
l2[2:-1]=[51,61,71]; print(l2)  # 元より長いシーケンスで置き換える
l2[2:-1]=[62]; print(l2)        # 元より短いシーケンスで置き換える
l2[1:1]=[23,24]; print(l2)      # シーケンスの挿入もできる（1の位置に挿入）
l2[1:3]=[]; print(l2)           # 部分シーケンスの削除もできる（1の位置 ～ 3の位置 の範囲を削除）
````

### リスト要素の追加・削除

- 上の「部分リストの置き換え」でも可能だが，専用のメソッドなども便利

#### `リスト.insert(位置, 値)`
- リストの指定した位置に値を挿入する

````python
l2 = [1,3,5,7,9]
l2.insert(1, 2)     # 1 の位置に 2 を挿入
print(l2)
l2.insert(-1, 8)    # -1 の位置に 8 を挿入
print(l2)
````

#### `リスト.append(値)`
- リストの末尾に値を挿入する

````python
l2 = [1,3,5,7,9]
l2.append(11)
print(l2)
````

#### `リスト.pop(位置)`
- リストの指定した位置の要素を削除し，その削除した値を返す
- `位置` を省略した場合，リストの末尾が対象となる

````python
l2 = [1,3,5,7,9]
print(l2.pop())
print(l2.pop(2))
print(l2.pop(0))
print(l2)
````

#### `del(リスト[位置])`
- 指定した位置の要素（または部分リスト）を削除する

````python
l2 = [1,3,5,7,9]
del(l2[2])          # 2番の要素を削除
print(l2)
del(l2[1:-1])       # 1番の位置 ～ −1番の位置 の範囲を削除
print(l2)
````

### for-in 文とシーケンス

- シーケンスの各要素に対して同じ処理を反復する場合，for-in 文を用いる．for-in 文は以下の形式を持つ
    <div style="border: solid gray 1px;">

    ```python
    for 変数… in シーケンス:
        反復される処理
        ...
    次の処理
    ```
    </div>

    注意：

    - `反復される処理 ...` の部分は，(`for`，`次の処理 ...` に対して) <u>**必ず4文字下げる**</u>．
    - `次の処理 ...` は，<u>**4文字戻して `for` とそろえる**</u>．
  

````python
for x in [1,3,5]:   # リストはシーケンスの一つ
    print(x)

print('---')        # 区切り表示

for x in (2,4,6):   # タプルもシーケンスの一つ
    print(x)

print('---')        # 区切り表示

for x in 'ABC':     # 文字列もシーケンスの一つ
    print(repr(x))
````

### シーケンスを生成する関数

- for-in 文は，シーケンスを生成する関数とともに用いられることも多い．

#### range 関数

- range 関数は整数値のシーケンス（range シーケンス）を生成する．
    - 生成するシーケンスは range という型のオブジェクトでリストではないが，list 関数でリストに変換できる
    - `range(m, n)` は，$m\ ～\ n-1$ の整数値の並びを返す
        - `m` を省略した `range(n)` は，$0\ ～\ n-1$ の整数値の並びを返す
    - 第3引数で増分を与えることもできる（省略時は $1$ とみなされている）
    - 負の増分を与えることで，減少する整数値のシーケンスが得られる
- range シーケンスは，そのまま for-in 文で使える

````python
print(range(1, 4))
print(list(range(1, 4)))        # 1以上4未満の整数を順に含むシーケンス
print(list(range(4)))           # 0以上4未満の整数を順に含むシーケンス
print(list(range(0, 4, 2)))     # 0以上4未満の偶数を順に含むシーケンス
print(list(range(4, 0, -1)))    # 4以下で0より大きい整数の降順のシーケンス
print(list(range(4, 0, -2)))    # 4以下で0より大きい偶数の降順のシーケンス

print('---')                    # 区切り表示

for x in range(4, 0, -1):
    print(x)
````

#### enumerate 関数

- enumerate 関数はシーケンスを引数に取り，そのシーケンスの各要素の番号と値のペア（2個組のタプル）のシーケンス（enumerate シーケンス）を返す
- enumerate シーケンスの各要素は，長さ2のタプルなので，for-in 文では2つの変数で受け取れる

````python
l3 = ['A', 'B', 'C']
print(enumerate(l3))
print(list(enumerate(l3)))

print('---')                    # 区切り表示

for i, x in enumerate(l3):      # l3 の各要素の番号を i，値を x で受け取る
    print(f'{i}番の要素は{x}')
````

#### [補足] zip 関数
- zip 関数は，同じ長さの複数のシーケンスを引数に取り，それらを組み合わせたタプルのシーケンスを返す

````python
l3 = ['A', 'B', 'C']
l4 = [65, 66, 67]
l5 = ['a', 'b', 'c']
print(zip(l3, l4))
print(list(zip(l3, l4)))
print(list(zip(l3, l4, l5)))

print('---')                    # 区切り表示

for X, c, x in zip(l3, l4, l5): 
    print(f'{X}のコードは{c}，小文字は{x}')
````

### 辞書とfor-in 文

- `辞書.keys()`，`辞書.values()`，`辞書.items()` で得られる値は，それぞれ dict_keys，dict_values，dict_items
    という型のシーケンスなので，for-in 文 で使用できる

````python
l3 = ['A', 'B', 'C']
l4 = [65, 66, 67]
d1 = dict(zip(l3,l4))   # 2個組タプルのシーケンスは dict 関数で辞書に変換できる
print(d1)
print(list(d1.keys()), list(d1.values()), list(d1.items()))

print('---')                        # 区切り表示

for k in d1.keys(): 
    print(f'キー{k}の値は{d1[k]}')  # d1 を使えばキーに対する値は得られる

print('---')                        # 区切り表示

for v in d1.values(): 
    print(f'値は{v}')               # キーを知る方法はない

print('---')                        # 区切り表示

for k, v in d1.items(): 
    print(f'キー{k}の値は{v}')      # d1 を使わなくてもキーと値が得られる
````

### for-in を用いたリストの表現（リストの内包表記）

- `[式 for 変数… in シーケンス]` という形で，規則的なリストを生成できる（for-in 文とはまた異なる）

````python
print([x for x in range(5)])    # range(5) の各値を並べたリスト
print([x*x for x in range(5)])  # range(5) の各値の2乗を並べたリスト
l1 = [1,4,2,8,5,7]
print([x*x for x in l1])        # l1 の各要素の2乗を並べたリスト
print([c for c in 'ABCD'])      # 文字列の各文字をバラバラに

print([f'{m}月' for m in range(1,13)])
print([f'{x}は{i+1}月' for i, x in enumerate(['睦月', '如月', '弥生'])])
````
